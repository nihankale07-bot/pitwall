```python
import pandas as pd
import numpy as np

from pathlib import Path

DATA_DIR = Path("../data/processed/9558")

lap_data = pd.read_parquet(DATA_DIR / "lap_performance.parquet")

print("Shape:", lap_data.shape)
print("\nColumns:")
print(lap_data.columns.tolist())

print("\nFirst 5 rows:")
display(lap_data.head())
```

```python
print("Tyre age summary:")
display(
    lap_data["tyre_age"].describe()
)

print("\nTyre age by compound:")
display(
    lap_data.groupby("compound")["tyre_age"]
    .agg(["count", "min", "median", "mean", "max"])
)

print("\nNumber of observations by compound:")
display(
    lap_data["compound"].value_counts()
)
```

```python
for compound in ["SOFT", "MEDIUM", "HARD"]:
    subset = lap_data[lap_data["compound"] == compound]

    print(f"\n{compound}")
    print("-" * 40)

    display(
        subset.groupby("tyre_age")["lap_duration"]
        .agg(["count", "mean", "median"])
        .head(20)
    )
```

```python
stint_check = (
    lap_data[
        lap_data["compound"].isin(["SOFT", "MEDIUM", "HARD"])
    ].sort_values(
        ["driver_number","stint_number","lap_number"]
    )
)

display(
    stint_check[
        [
            "driver_number",
            "stint_number",
            "lap_number",
            "compound",
            "tyre_age",
            "lap_duration"
        ]
    ].head(40)
)
```

```python
stint_progression = (
    stint_check.groupby(
        ["driver_number","stint_number","compound"]
    )
    .agg(
        laps=("lap_number", "count"),
        first_lap=("lap_number", "min"),
        last_lap=("lap_number", "max"),
        min_tyre_age=("tyre_age", "min"),
        max_tyre_age=("tyre_age", "max")
    ).reset_index()
)
display(stint_progression.head(30))
```

```python
degradation_data = lap_data[
    lap_data["compound"].isin(["SOFT","MEDIUM","HARD"])
].copy()

degradation_data = degradation_data[
    degradation_data["tyre_age"] >= 1
].copy()

print("Candidate degradation dataset shape:")
print(degradation_data.shape)
print("\nRows by compound:")
print(
    degradation_data["compound"].value_counts()
)

print("\nTyre-age coverage:")
display(
    degradation_data
    .groupby("compound")["tyre_age"]
    .agg(
        count="count",
        min="min",
        median="median",
        mean="mean",
        max="max"
    )
)
```

```python
degradation_summary = (
    degradation_data.groupby(["compound", "tyre_age"])["lap_duration"]
    .agg(
        count="count",
        mean="mean",
        median="median",
        std="std"
    ).reset_index()
)

for compound in ["SOFT", "MEDIUM", "HARD"]:
    print(f"{compound}")

    display(
        degradation_summary[
            (degradation_summary["compound"] == compound) & (degradation_summary["count"] >= 3)
        ]
    )
```

```python
fuel_features = pd.read_parquet(
    DATA_DIR / "fuel_features.parquet"
)

print("Fuel featues:")
print(fuel_features.head())

print(fuel_features.shape)

degradation_data = degradation_data.merge(
    fuel_features,
    on=["driver_number", "lap_number"],
    how="left",
    validate="one_to_one"
)

print("Merged shape:", degradation_data.shape)
print("Missing fuel effects:")
print(degradation_data["fuel_effect"].isna().sum())

display(
    degradation_data[
        [
            "driver_number",
            "lap_number",
            "compound",
            "tyre_age",
            "lap_duration",
            "fuel_effect"
        ]
    ].head(20)
)
```

```python
degradation_data["fuel_adjusted_lap_duration"] = (
    degradation_data["lap_duration"] - degradation_data["fuel_effect"]
)

display(
    degradation_data[
        [
            "driver_number",
            "lap_number",
            "compound",
            "tyre_age",
            "lap_duration",
            "fuel_effect",
            "fuel_adjusted_lap_duration"
        ]
    ].head(20)
)
```

```python
adjusted_degradation_summary = (
    degradation_data
    .groupby(["compound","tyre_age"])["fuel_adjusted_lap_duration"]
    .agg(
        count="count",
        mean="mean",
        median="median",
        std="std"
    ).reset_index()
)
for compound in ["SOFT", "MEDIUM", "HARD"]:
    print(f"{compound} - FUEL ADJUSTED")
    display(
        adjusted_degradation_summary[
            (adjusted_degradation_summary["compound"] == compound) &
            (adjusted_degradation_summary["count"] >= 3)
        ]
    )
```

```python

for compound in ["SOFT", "MEDIUM", "HARD"]:

    subset = degradation_data[
        degradation_data["compound"] == compound
    ]["fuel_adjusted_lap_duration"]

    q1 = subset.quantile(0.25)
    q3 = subset.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = subset[
        (subset < lower) | (subset > upper)
    ]

    print(f"\n===== {compound} =====")
    print(f"Q1: {q1:.3f}")
    print(f"Q3: {q3:.3f}")
    print(f"IQR: {iqr:.3f}")
    print(f"Lower bound: {lower:.3f}")
    print(f"Upper bound: {upper:.3f}")
    print(f"Potential outliers: {len(outliers)}")
    print(f"Total observations: {len(subset)}")
```

```python
for compound in ["SOFT","MEDIUM", "HARD"]:
    subset = degradation_data[
        degradation_data["compound"] == compound
    ].copy()

    q1 = subset["fuel_adjusted_lap_duration"].quantile(0.25)
    q3 = subset["fuel_adjusted_lap_duration"].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5*iqr
    upper = q3 + 1.5*iqr
    suspicious = subset[
        (subset["fuel_adjusted_lap_duration"] < lower) |
        (subset["fuel_adjusted_lap_duration"] > upper)
    ]

    print(f"\n===== {compound} suspicious laps =====")

    display(
        suspicious[
            [
                "driver_number",
                "lap_number",
                "stint_number",
                "compound",
                "tyre_age",
                "lap_duration",
                "fuel_effect",
                "fuel_adjusted_lap_duration"
            ]
        ]
        .sort_values("fuel_adjusted_lap_duration", ascending=False)
    )
```

```python
tyre_model_data = degradation_data[
    ~degradation_data["low_grip_condition"].eq(1)
].copy()
print("Original degradation data:", degradation_data.shape)
print("Tyre model data:", tyre_model_data.shape)
print(
    tyre_model_data["low_grip_condition"].value_counts()
)
```

```python
tyre_age_summary = (
    tyre_model_data
    .groupby(["compound", "tyre_age"])["fuel_adjusted_lap_duration"]
    .agg(["count","mean","median"])
    .reset_index()
)
print(tyre_age_summary.to_string(index=False))
```

```python
import statsmodels.formula.api as smf

tyre_baseline_model = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) + tyre_age",
    data=tyre_model_data
).fit()

print(tyre_baseline_model.summary())
```

```python
tyre_degradation_nonlinear = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) + tyre_age + I(tyre_age**2)",
    data=tyre_model_data
).fit()

print(tyre_degradation_nonlinear.summary())
```

```python
import numpy as np
import pandas as pd

ages = np.arange(
    tyre_model_data["tyre_age"].min(),
    tyre_model_data["tyre_age"].max() + 1
)

curve = pd.DataFrame({
    "tyre_age": ages
})

curve["predicted_degradation"] = (
    -0.2144 * curve["tyre_age"]
    + 0.0182 * curve["tyre_age"] ** 2
)

print(curve)
```

```python
tyre_compound_model = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) * (tyre_age + I(tyre_age**2))",
    data=tyre_model_data
).fit()

print(tyre_compound_model.summary())
```

```python

split_lap = 45

train_data = tyre_model_data[
    tyre_model_data["lap_number"] < split_lap
].copy()

test_data = tyre_model_data[
    tyre_model_data["lap_number"] >= split_lap
].copy()

print("Training rows:", len(train_data))
print("Testing rows:", len(test_data))

print(
    "Train lap range:",
    train_data["lap_number"].min(),
    "-",
    train_data["lap_number"].max()
)

print(
    "Test lap range:",
    test_data["lap_number"].min(),
    "-",
    test_data["lap_number"].max()
)
```

```python
import statsmodels.formula.api as smf
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

validation_data = tyre_model_data.sort_values("lap_number").copy()

print("Training rows:", len(train_data))
print("Testing rows:", len(test_data))
print("Train lap range:", train_data["lap_number"].min(), "-", train_data["lap_number"].max())
print("Test lap range:", test_data["lap_number"].min(), "-", test_data["lap_number"].max())

linear_model = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) + tyre_age",
    data=train_data
).fit()


quadratic_model = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) + tyre_age + I(tyre_age**2)",
    data=train_data
).fit()

linear_pred = linear_model.predict(test_data)
quadratic_pred = quadratic_model.predict(test_data)

actual = test_data["fuel_adjusted_lap_duration"]

linear_r2 = r2_score(actual, linear_pred)
quadratic_r2 = r2_score(actual, quadratic_pred)

linear_rmse = np.sqrt(mean_squared_error(actual, linear_pred))
quadratic_rmse = np.sqrt(mean_squared_error(actual, quadratic_pred))


print("\n===== TEST SET RESULTS =====")

print("\nLinear model:")
print("R²:", round(linear_r2, 3))
print("RMSE:", round(linear_rmse, 3), "seconds")

print("\nQuadratic model:")
print("R²:", round(quadratic_r2, 3))
print("RMSE:", round(quadratic_rmse, 3), "seconds")
```

```python
tyre_piecewise_data = tyre_model_data.copy()

tyre_piecewise_data["tyre_age_phase"] = pd.cut(
    tyre_piecewise_data["tyre_age"],
    bins=[0,5,12,float("inf")],
    labels=["early", "middle", "late"],
    include_lowest=True
)

print(
    tyre_piecewise_data[
        ["tyre_age","tyre_age_phase"]
    ].drop_duplicates().sort_values("tyre_age").to_string(index=False)
)
print("Phase counts:", tyre_piecewise_data["tyre_age_phase"].value_counts().sort_index())
```

```python
import statsmodels.formula.api as smf
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

train_data = tyre_piecewise_data[
    tyre_piecewise_data["lap_number"] < 45
].copy()

test_data = tyre_piecewise_data[
    tyre_piecewise_data["lap_number"] >= 45
].copy()

linear_model = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) + tyre_age",
    data=train_data
).fit()

piecewise_model = smf.ols(
    "fuel_adjusted_lap_duration ~ C(driver_number) + C(compound) + tyre_age + C(tyre_age_phase)",
    data=train_data
).fit()

actual = test_data["fuel_adjusted_lap_duration"]

linear_pred = linear_model.predict(test_data)
piecewise_model = piecewise_model.predict(test_data)

linear_r2 = r2_score(actual, linear_pred)
linear_rmse = np.sqrt(mean_squared_error(actual, linear_pred))

piecewise_r2 = r2_score(actual, piecewise_model)
piecewise_rmse = np.sqrt(mean_squared_error(actual, piecewise_model))

print("\nLinear model:")
print("R²:", round(linear_r2, 3))
print("RMSE:", round(linear_rmse, 3), "seconds")

print("\nPiecewise model:")
print("R²:", round(piecewise_r2, 3))
print("RMSE:", round(piecewise_rmse, 3), "seconds")


# --------------------------------------------------
# PIECEWISE MODEL COEFFICIENTS
# --------------------------------------------------

print("\n===== PIECEWISE COEFFICIENTS =====")

print(
    piecewise_model.params[
        piecewise_model.params.index.str.contains("tyre_age_phase")
    ]
)
```
