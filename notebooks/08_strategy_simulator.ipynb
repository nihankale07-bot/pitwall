```python

# Cell 8: Assemble the simulator's race-state table

keys = ["driver_number", "lap_number"]

# Start with the existing race-state dataset
sim_state = race_state.copy()

# Merge previously engineered features
feature_tables = [
    fuel_features[["driver_number", "lap_number", "fuel_effect"]],
    traffic_features[
        ["driver_number", "lap_number",
         "traffic_ahead", "traffic_behind", "position_change"]
    ].drop_duplicates(subset=keys),
    tyre_features[
        ["driver_number", "lap_number", "tyre_degradation_effect"]
    ],
]

for features in feature_tables:
    sim_state = sim_state.merge(
        features,
        on=keys,
        how="left",
        validate="one_to_one"
    )

# Use the existing race-state tyre age where available.
# The engineered tyre age is only a fallback.
if "tyre_age" in sim_state.columns:
    if "tyre_age_feature" not in sim_state.columns:
        # race_state already contains tyre_age
        pass

# Sort for reliable race progression
sim_state = sim_state.sort_values(keys).reset_index(drop=True)

print("Simulator race-state shape:", sim_state.shape)
print("Drivers:", sim_state["driver_number"].nunique())
print("Lap range:", sim_state["lap_number"].min(), "to",
      sim_state["lap_number"].max())
print("Unique driver-lap keys:",
      sim_state[keys].drop_duplicates().shape[0])
print("Duplicate driver-lap keys:", sim_state.duplicated(keys).sum())

display(
    sim_state[
        ["driver_number", "lap_number", "position", "compound",
         "tyre_age", "lap_duration", "fuel_effect",
         "traffic_ahead", "traffic_behind", "rainfall"]
    ].head(10)
)
```

```python
race_results = []

for driver, group in sim_state.groupby("driver_number"):
    group = group.sort_values("lap_number")

    valid_times = group["lap_duration"].dropna()
    valid_positions = group["position"].dropna()

    compounds = (
        group.dropna(subset=["compound"])
        .sort_values("lap_number")
        .groupby("stint_number", dropna=False)["compound"]
        .first()
        .tolist()
    )

    race_results.append({
        "driver_number": driver,
        "completed_laps": group["lap_number"].max(),
        "last_recorded_position": (
            valid_positions.iloc[-1] if not valid_positions.empty else None
        ),
        "total_recorded_lap_time_s": valid_times.sum(),
        "valid_lap_times": len(valid_times),
        "observed_stint_sequence": " -> ".join(map(str, compounds)),
        "stints_recorded": len(compounds),
    })

race_results = pd.DataFrame(race_results)

race_results = race_results.sort_values(
    ["last_recorded_position", "driver_number"],
    na_position="last"
).reset_index(drop=True)

display(race_results.round(3))
```

```python

# Cell 13: Build an all-driver snapshot for any race lap

def get_race_snapshot(data, lap_number):
    snapshot = data.loc[
        data["lap_number"] == lap_number
    ].copy()

    snapshot = snapshot.sort_values(
        ["position", "driver_number"],
        na_position="last"
    ).reset_index(drop=True)

    # Add readable names for the drivers in this race
    driver_names = {
        1: "Max Verstappen",
        2: "Logan Sargeant",
        4: "Lando Norris",
        10: "Pierre Gasly",
        11: "Sergio Perez",
        14: "Fernando Alonso",
        16: "Charles Leclerc",
        18: "Lance Stroll",
        20: "Kevin Magnussen",
        22: "Yuki Tsunoda",
        23: "Alexander Albon",
        24: "Guanyu Zhou",
        27: "Nico Hulkenberg",
        31: "Esteban Ocon",
        44: "Lewis Hamilton",
        55: "Carlos Sainz",
        63: "George Russell",
        77: "Valtteri Bottas",
        81: "Oscar Piastri",
    }

    snapshot["driver_name"] = snapshot["driver_number"].map(
        driver_names
    ).fillna("Unknown")

    columns = [
        "position",
        "driver_number",
        "driver_name",
        "lap_number",
        "lap_duration",
        "compound",
        "tyre_age",
        "stint_number",
        "rainfall",
        "traffic_ahead",
        "traffic_behind",
    ]

    columns = [c for c in columns if c in snapshot.columns]

    return snapshot[columns]


# Inspect the race state immediately before the wet-weather pit window
snapshot_lap = 26

race_snapshot = get_race_snapshot(sim_state, snapshot_lap)

print(f"Race state after recorded lap {snapshot_lap}")
print("Drivers represented:", len(race_snapshot))
display(race_snapshot)
```

```python

# Cell 14: Define candidate strategies for George Russell

TARGET_DRIVER = 63
DECISION_LAP = 26
RACE_LAPS = int(sim_state["lap_number"].max())

# Actual first recorded lap on intermediates for George
george_data = sim_state[
    sim_state["driver_number"] == TARGET_DRIVER
].sort_values("lap_number")

actual_intermediate_laps = george_data.loc[
    george_data["compound"] == "INTERMEDIATE",
    "lap_number"
]

if actual_intermediate_laps.empty:
    raise ValueError("No observed intermediate stint found for George.")

ACTUAL_INTERMEDIATE_START = int(actual_intermediate_laps.min())

# Candidate first laps on intermediates.
# These are scenarios to test, not predictions of the best strategy.
candidate_laps = sorted(set([
    max(1, ACTUAL_INTERMEDIATE_START - 2),
    max(1, ACTUAL_INTERMEDIATE_START - 1),
    ACTUAL_INTERMEDIATE_START,
    ACTUAL_INTERMEDIATE_START + 1,
    ACTUAL_INTERMEDIATE_START + 2,
]))

candidate_strategies = pd.DataFrame({
    "strategy_id": [
        f"INTERS_LAP_{lap}" for lap in candidate_laps
    ],
    "first_intermediate_lap": candidate_laps,
})

candidate_strategies["decision_lap"] = (
    candidate_strategies["first_intermediate_lap"] - 1
)

candidate_strategies["laps_remaining_at_decision"] = (
    RACE_LAPS - candidate_strategies["decision_lap"]
)

candidate_strategies["is_historical_timing"] = (
    candidate_strategies["first_intermediate_lap"]
    == ACTUAL_INTERMEDIATE_START
)

print("Race length:", RACE_LAPS, "laps")
print("George's observed first intermediate lap:",
      ACTUAL_INTERMEDIATE_START)
print("\nCandidate strategy scenarios:")
display(candidate_strategies)

# Verify George's position and weather at each proposed decision lap
decision_context = george_data[
    george_data["lap_number"].isin(
        candidate_strategies["decision_lap"]
    )
][[
    "lap_number",
    "position",
    "compound",
    "tyre_age",
    "lap_duration",
    "rainfall",
    "track_temperature",
]].copy()

print("\nHistorical context at candidate decision laps:")
display(decision_context.sort_values("lap_number").reset_index(drop=True))
```

```python
pace_data = sim_state.copy()

# Keep valid timed laps with known tyre compounds
pace_data = pace_data[
    pace_data["lap_duration"].notna()
    & pace_data["compound"].notna()
].copy()

# Exclude pit-out laps from representative running pace
if "is_pit_out_lap" in pace_data.columns:
    pace_data = pace_data[
        ~pace_data["is_pit_out_lap"].fillna(False)
    ].copy()

# Focus on the changing-weather phase around the strategy window
pace_data["race_phase"] = pd.cut(
    pace_data["lap_number"],
    bins=[0, 20, 26, 38, 52],
    labels=["Laps 1-20", "Laps 21-26", "Laps 27-38", "Laps 39-52"]
)

pace_summary = (
    pace_data.groupby(
        ["race_phase", "compound", "rainfall"],
        observed=True
    )
    .agg(
        lap_samples=("lap_duration", "size"),
        median_lap_time=("lap_duration", "median"),
        mean_lap_time=("lap_duration", "mean"),
        median_tyre_age=("tyre_age", "median"),
    )
    .reset_index()
    .sort_values(["race_phase", "compound"])
)

print("Observed pace by race phase, compound, and rainfall")
display(pace_summary.round(3))

print("\nSamples available for each compound:")
display(
    pace_data.groupby("compound", observed=True)
    .agg(
        valid_laps=("lap_duration", "size"),
        drivers=("driver_number", "nunique"),
        median_lap_time=("lap_duration", "median"),
    )
    .reset_index()
    .round(3)
)
```

```python
# Cell 16: Build lap-by-lap tyre pace references

pace_window = sim_state[
    sim_state["lap_number"].between(20, 35)
    & sim_state["lap_duration"].notna()
    & sim_state["compound"].notna()
].copy()

# Exclude pit-out laps from representative running pace
if "is_pit_out_lap" in pace_window.columns:
    pace_window = pace_window[
        ~pace_window["is_pit_out_lap"].fillna(False)
    ].copy()

lap_pace_reference = (
    pace_window
    .groupby(
        ["lap_number", "compound", "rainfall"],
        observed=True
    )
    .agg(
        drivers_observed=("driver_number", "nunique"),
        median_lap_time=("lap_duration", "median"),
        median_tyre_age=("tyre_age", "median"),
    )
    .reset_index()
    .sort_values(["lap_number", "rainfall", "compound"])
)

print("Lap-by-lap compound pace references, laps 20-35")
display(lap_pace_reference.round(3))

print("\nCoverage by lap and compound")
display(
    lap_pace_reference.pivot_table(
        index="lap_number",
        columns="compound",
        values="drivers_observed",
        aggfunc="first",
        fill_value=0
    )
)
```

```python
# Cell 17: Estimate observed tyre crossover

crossover = lap_pace_reference[
    lap_pace_reference["compound"].isin(["MEDIUM", "INTERMEDIATE"])
].pivot(
    index="lap_number",
    columns="compound",
    values="median_lap_time"
).reset_index()

# Keep only laps where both compounds have observed pace
crossover = crossover.dropna(
    subset=["MEDIUM", "INTERMEDIATE"]
).copy()

crossover["intermediate_minus_medium_sec"] = (
    crossover["INTERMEDIATE"] - crossover["MEDIUM"]
)

crossover["cumulative_gap_sec"] = (
    crossover["intermediate_minus_medium_sec"].cumsum()
)

print("Observed same-lap tyre pace gap")
display(crossover.round(3))

print(
    "\nInterpretation: positive gap = intermediates were slower; "
    "negative gap = intermediates were faster."
)
```

```python
# Cell 18: Inspect observed tyre changes and surrounding lap times

transition_data = sim_state.sort_values(
    ["driver_number", "lap_number"]
).copy()

transition_data["previous_compound"] = (
    transition_data.groupby("driver_number")["compound"].shift(1)
)

transition_data["previous_lap_number"] = (
    transition_data.groupby("driver_number")["lap_number"].shift(1)
)

transition_data["compound_changed"] = (
    transition_data["compound"].notna()
    & transition_data["previous_compound"].notna()
    & transition_data["compound"].ne(
        transition_data["previous_compound"]
    )
)

transitions = transition_data[
    transition_data["compound_changed"]
].copy()

transitions["driver_lap_label"] = (
    transitions["driver_number"].astype(str)
    + " | lap "
    + transitions["lap_number"].astype(str)
)

print("Observed compound changes")
display(
    transitions[
        [
            "driver_number",
            "previous_lap_number",
            "lap_number",
            "previous_compound",
            "compound",
            "lap_duration",
            "is_pit_out_lap",
            "position",
        ]
    ]
    .sort_values(["lap_number", "driver_number"])
    .round(3)
)

# Inspect laps immediately around each driver's compound change
near_transition = transition_data[
    transition_data.apply(
        lambda row: any(
            (
                transitions["driver_number"].eq(row["driver_number"])
                & (transitions["lap_number"] - row["lap_number"]).abs().le(1)
            )
        ),
        axis=1,
    )
].copy()

print("\nLap times around observed compound changes")
display(
    near_transition[
        [
            "driver_number",
            "lap_number",
            "compound",
            "lap_duration",
            "is_pit_out_lap",
            "position",
        ]
    ]
    .sort_values(["driver_number", "lap_number"])
    .round(3)
)
```

```python
# Cell 19: Estimate pit-out lap excess versus same-lap running pace

pit_data = sim_state[
    sim_state["lap_duration"].notna()
    & sim_state["compound"].notna()
].copy()

pit_data["is_pit_out_lap"] = (
    pit_data["is_pit_out_lap"].fillna(False).astype(bool)
)

# Reference pace: non-pit-out laps on the same lap and compound
running_reference = (
    pit_data[~pit_data["is_pit_out_lap"]]
    .groupby(["lap_number", "compound"], observed=True)
    .agg(
        running_median=("lap_duration", "median"),
        running_drivers=("driver_number", "nunique"),
    )
    .reset_index()
)

pit_out_laps = pit_data[pit_data["is_pit_out_lap"]].merge(
    running_reference,
    on=["lap_number", "compound"],
    how="left",
    validate="many_to_one",
)

pit_out_laps["pit_out_excess_sec"] = (
    pit_out_laps["lap_duration"] - pit_out_laps["running_median"]
)

print("Pit-out laps compared with same-lap, same-compound running pace")
display(
    pit_out_laps[
        [
            "driver_number",
            "lap_number",
            "compound",
            "lap_duration",
            "running_median",
            "running_drivers",
            "pit_out_excess_sec",
        ]
    ]
    .sort_values(["lap_number", "driver_number"])
    .round(3)
)

print("\nSummary of usable pit-out excess estimates")
display(
    pit_out_laps["pit_out_excess_sec"]
    .dropna()
    .describe()
    .round(3)
    .to_frame("pit_out_excess_sec")
)
```

```python
# Cell 20: Strategy sensitivity analysis for George Russell

import numpy as np
import pandas as pd

# Observed same-lap pace differences from Cell 17
observed_gaps = crossover.set_index("lap_number")[
    "intermediate_minus_medium_sec"
].to_dict()

# Scenarios for the missing counterfactual:
# positive = mediums are slower than intermediates after lap 27
# These are assumptions to test, not learned facts.
post_crossover_scenarios = {
    "Mediums equal to inters": 0.0,
    "Mediums 2s slower per lap": 2.0,
    "Mediums 5s slower per lap": 5.0,
}

candidate_laps = [26, 27, 28, 29, 30]
last_lap = int(sim_state["lap_number"].max())

results = []

for scenario_name, medium_disadvantage in post_crossover_scenarios.items():
    for pit_lap in candidate_laps:
        total_delta = 0.0
        unsupported_laps = 0

        for lap in range(21, last_lap + 1):
            if lap < pit_lap:
                # George stays on mediums
                if lap in observed_gaps:
                    total_delta += 0.0
                elif lap > 27:
                    # No reliable observed medium/intermediate comparison
                    total_delta += 0.0
                    unsupported_laps += 1

            else:
                # George has switched to intermediates
                if lap in observed_gaps:
                    total_delta += observed_gaps[lap]
                elif lap > 27:
                    # Scenario assumption for the unobserved medium alternative
                    total_delta -= medium_disadvantage
                    unsupported_laps += 1

        results.append({
            "scenario": scenario_name,
            "intermediate_start_lap": pit_lap,
            "assumed_medium_disadvantage_sec_per_lap": medium_disadvantage,
            "estimated_delta_sec": round(total_delta, 2),
            "unsupported_laps": unsupported_laps,
        })

strategy_sensitivity = pd.DataFrame(results)

print("Sensitivity results (exploratory, not a validated race prediction)")
display(
    strategy_sensitivity.pivot(
        index="intermediate_start_lap",
        columns="scenario",
        values="estimated_delta_sec",
    ).round(2)
)

print(
    "\nImportant: these totals are scenario comparisons, not predicted finishing "
    "times or win probabilities. The unsupported-lap count shows where "
    "counterfactual pace assumptions enter the calculation."
)
```

```python
def build_decision_state(sim_state, driver_number, decision_lap):
    required_columns = [
        "driver_number",
        "lap_number",
        "position",
        "compound",
        "tyre_age",
    ]

    missing_columns = [
        col for col in required_columns
        if col not in sim_state.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Missing required columns: {missing_columns}"
        )
    if decision_lap < 1:
        raise ValueError("decision_lap must be at least 1.")

    # Latest available state for every driver at this lap
    field_state = (
        sim_state.loc[
            sim_state["lap_number"].eq(decision_lap)
        ]
        .sort_values("position", na_position="last")
        .copy()
        .reset_index(drop=True)
    )

    if field_state.empty:
        raise ValueError(
            f"No race-state data available for lap {decision_lap}."
        )

    # Find the selected driver
    target_rows = field_state.loc[
        field_state["driver_number"].eq(driver_number)
    ]

    if target_rows.empty:
        available_drivers = sorted(
            field_state["driver_number"].dropna().unique().tolist()
        )

        raise ValueError(
            f"Driver {driver_number} has no recorded state "
            f"on lap {decision_lap}. "
            f"Drivers available: {available_drivers}"
        )

    target_state = target_rows.iloc[0].copy()

    # Automatically collect the useful state fields already available
    context_columns = [
        "driver_number",
        "lap_number",
        "position",
        "compound",
        "tyre_age",
        "stint_number",
        "lap_duration",
        "is_pit_out_lap",
        "fuel_effect",
        "tyre_degradation_effect",
        "traffic_ahead",
        "traffic_behind",
        "position_change",
        "rainfall",
        "humidity",
        "air_temperature",
        "track_temperature",
        "wind_speed",
        "grip_condition",
        "blue_flag",
    ]

    context_columns = [
        col for col in context_columns
        if col in field_state.columns
    ]

    field_context = field_state[context_columns].copy()

    # Flag missing information for the selected driver
    important_fields = [
        "position",
        "compound",
        "tyre_age",
        "fuel_effect",
        "tyre_degradation_effect",
        "traffic_ahead",
        "traffic_behind",
        "rainfall",
        "air_temperature",
    ]

    important_fields = [
        col for col in important_fields
        if col in target_state.index
    ]

    missing_features = [
        col for col in important_fields
        if pd.isna(target_state[col])
    ]

    decision_context = {
        "driver_number": int(driver_number),
        "decision_lap": int(decision_lap),
        "next_lap": int(decision_lap + 1),
        "race_laps_in_dataset": int(
            sim_state["lap_number"].max()
        ),
        "drivers_observed": int(
            field_state["driver_number"].nunique()
        ),
        "target_state": target_state,
        "field_state": field_context,
        "missing_target_features": missing_features,
    }

    return decision_context


# First test: George Russell after lap 26
decision_state = build_decision_state(
    sim_state=sim_state,
    driver_number=63,
    decision_lap=26,
)

print("PITWALL DECISION STATE")
print("=" * 45)
print("Selected driver:", decision_state["driver_number"])
print("State after lap:", decision_state["decision_lap"])
print("Next lap to strategize:", decision_state["next_lap"])
print("Drivers observed:", decision_state["drivers_observed"])
print(
    "Missing important features:",
    decision_state["missing_target_features"]
)

print("\nSELECTED DRIVER STATE")
display(
    decision_state["target_state"]
    .to_frame("value")
)

print("\nFIELD STATE")
display(decision_state["field_state"])
```

```python


def generate_strategy_candidates(
    decision_state,
    sim_state,
    pit_window=5,
):
    target = decision_state["target_state"]
    decision_lap = decision_state["decision_lap"]
    next_lap = decision_state["next_lap"]
    driver_number = decision_state["driver_number"]

    current_compound = target["compound"]

    if pd.isna(current_compound):
        raise ValueError(
            f"Driver {driver_number} has no recorded compound "
            f"at lap {decision_lap}."
        )

    current_compound = str(current_compound).upper()

    last_race_lap = int(sim_state["lap_number"].max())

    # Candidate pit laps must be after the completed decision lap
    candidate_pit_laps = list(
        range(
            next_lap,
            min(decision_lap + pit_window, last_race_lap - 1) + 1,
        )
    )

    # Compounds actually observed around this decision window
    observed_window = sim_state[
        sim_state["lap_number"].between(
            max(1, decision_lap - 2),
            min(last_race_lap, decision_lap + pit_window),
        )
        & sim_state["compound"].notna()
    ].copy()

    available_compounds = sorted(
        observed_window["compound"]
        .astype(str)
        .str.upper()
        .unique()
        .tolist()
    )

    # Use observed weather to prioritize sensible compound choices.
    rainfall = target.get("rainfall", np.nan)
    grip = str(target.get("grip_condition", "")).upper()

    wet_conditions = (
        (pd.notna(rainfall) and float(rainfall) > 0)
        or grip in {"LOW", "WET"}
    )

    wet_compounds = {"INTERMEDIATE", "WET"}
    dry_compounds = {"SOFT", "MEDIUM", "HARD"}

    if wet_conditions:
        eligible_compounds = [
            compound for compound in available_compounds
            if compound in wet_compounds
        ]
    else:
        eligible_compounds = [
            compound for compound in available_compounds
            if compound in dry_compounds
        ]

    # Always include the option to remain on the current tyres
    # through the candidate window as a baseline.
    candidates = [{
        "driver_number": driver_number,
        "decision_lap": decision_lap,
        "strategy_type": "HOLD",
        "pit_lap": None,
        "target_compound": current_compound,
        "description": (
            f"Stay on {current_compound} through the candidate window"
        ),
    }]

    # Generate all feasible pit-lap / eligible-compound combinations
    for pit_lap in candidate_pit_laps:
        for compound in eligible_compounds:
            if compound == current_compound:
                continue

            candidates.append({
                "driver_number": driver_number,
                "decision_lap": decision_lap,
                "strategy_type": "PIT",
                "pit_lap": pit_lap,
                "target_compound": compound,
                "description": (
                    f"Switch to {compound} on lap {pit_lap}"
                ),
            })

    candidates_df = pd.DataFrame(candidates)

    if candidates_df.empty:
        raise ValueError(
            "No feasible strategy candidates were generated. "
            "Check compound availability and the decision lap."
        )

    candidates_df.insert(
        0,
        "strategy_id",
        [
            f"S{i + 1:03d}"
            for i in range(len(candidates_df))
        ],
    )

    candidates_df["is_hold_baseline"] = (
        candidates_df["strategy_type"] == "HOLD"
    )

    return candidates_df


# Generate strategies for George after lap 26
strategy_candidates = generate_strategy_candidates(
    decision_state=decision_state,
    sim_state=sim_state,
    pit_window=5,
)

print("Selected driver:", decision_state["driver_number"])
print("Decision lap:", decision_state["decision_lap"])
print("Candidate strategies:", len(strategy_candidates))

display(strategy_candidates)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Exclude pit-out laps so the model learns racing pace, not pit-exit delays
model_data = pace_data.loc[
    ~pace_data["is_pit_out_lap"]
    & pace_data["lap_duration"].notna()
    & pace_data["compound"].notna()
].copy()

# Keep only features available in the current simulation dataset
numeric_features = [
    "lap_number",
    "tyre_age",
    "rainfall",
    "humidity",
    "air_temperature",
    "track_temperature",
    "wind_speed",
    "traffic_ahead",
    "traffic_behind",
    "position_change",
]

categorical_features = ["compound", "driver_number"]

numeric_features = [
    col for col in numeric_features if col in model_data.columns
]
categorical_features = [
    col for col in categorical_features if col in model_data.columns
]

feature_columns = numeric_features + categorical_features

# Time-ordered split: earlier laps train, later laps test
split_lap = 35

train_data = model_data.loc[
    model_data["lap_number"] <= split_lap
].copy()

test_data = model_data.loc[
    model_data["lap_number"] > split_lap
].copy()

if train_data.empty or test_data.empty:
    raise ValueError("The temporal train/test split produced an empty dataset.")

X_train = train_data[feature_columns]
y_train = train_data["lap_duration"]

X_test = test_data[feature_columns]
y_test = test_data["lap_duration"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features),
])

pace_baseline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=300,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1,
    )),
])

pace_baseline.fit(X_train, y_train)
test_predictions = pace_baseline.predict(X_test)

mae = mean_absolute_error(y_test, test_predictions)
rmse = np.sqrt(mean_squared_error(y_test, test_predictions))
r2 = r2_score(y_test, test_predictions)

# Simple reference baseline: always predict the training-set median
median_baseline_prediction = np.full(
    len(y_test), y_train.median()
)
median_baseline_mae = mean_absolute_error(
    y_test, median_baseline_prediction
)

print("TIME-ORDERED PACE VALIDATION")
print("=" * 55)
print(f"Training laps: 1 to {split_lap}")
print(f"Test laps: {split_lap + 1} to {int(test_data['lap_number'].max())}")
print(f"Training rows: {len(train_data)}")
print(f"Test rows: {len(test_data)}")
print(f"Model MAE: {mae:.3f} seconds")
print(f"Model RMSE: {rmse:.3f} seconds")
print(f"Model R²: {r2:.3f}")
print(f"Median-baseline MAE: {median_baseline_mae:.3f} seconds")

# Check performance separately for each compound in the test period
evaluation = test_data[
    ["driver_number", "lap_number", "compound", "lap_duration"]
].copy()

evaluation["predicted_lap_duration"] = test_predictions
evaluation["absolute_error"] = (
    evaluation["lap_duration"]
    - evaluation["predicted_lap_duration"]
).abs()

print("\nTEST PERFORMANCE BY COMPOUND")
print("=" * 55)

compound_evaluation = (
    evaluation.groupby("compound")
    .agg(
        test_laps=("lap_duration", "size"),
        MAE=("absolute_error", "mean"),
        actual_median=("lap_duration", "median"),
        predicted_median=("predicted_lap_duration", "median"),
    )
    .round(3)
)

display(compound_evaluation)
```

```python

# Cell 26: Compare Linear Regression, Random Forest, and XGBoost

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from xgboost import XGBRegressor

# ---------------------------------------------------------
# 1. Build a clean dataset without pit-out laps
# ---------------------------------------------------------
model_data = sim_state.loc[
    sim_state["lap_duration"].notna()
    & sim_state["compound"].notna()
].copy()

model_data["compound"] = model_data["compound"].astype(str).str.upper()
model_data["driver_number"] = model_data["driver_number"].astype(str)

model_data["is_pit_out_lap"] = (
    model_data["is_pit_out_lap"].fillna(False).astype(bool)
)

model_data = model_data.loc[
    ~model_data["is_pit_out_lap"]
].copy()

model_data = model_data.sort_values(
    ["lap_number", "driver_number"]
).reset_index(drop=True)

# ---------------------------------------------------------
# 2. Create track evolution using weather data only
# ---------------------------------------------------------
# One weather record per lap. These are observed conditions,
# not lap-time outcomes.
weather_by_lap = (
    model_data.groupby("lap_number")
    .agg(
        rainfall=("rainfall", "median"),
        track_temperature=("track_temperature", "median"),
        air_temperature=("air_temperature", "median"),
        humidity=("humidity", "median"),
    )
    .sort_index()
)

# Rolling change over the previous 3 lap observations.
# shift(1) ensures the trend feature does not use the
# current lap's weather or any future observations.
for col in [
    "rainfall",
    "track_temperature",
    "air_temperature",
    "humidity",
]:
    previous_weather = weather_by_lap[col].shift(1)

    weather_by_lap[f"{col}_change_3lap"] = (
        previous_weather - previous_weather.shift(3)
    ) / 3

weather_features = [
    "rainfall_change_3lap",
    "track_temperature_change_3lap",
    "air_temperature_change_3lap",
    "humidity_change_3lap",
]

weather_by_lap = weather_by_lap.rename(columns={
    "rainfall_change_3lap": "rainfall_trend",
    "track_temperature_change_3lap": "track_temp_trend",
    "air_temperature_change_3lap": "air_temp_trend",
    "humidity_change_3lap": "humidity_trend",
})

weather_features = [
    "rainfall_trend",
    "track_temp_trend",
    "air_temp_trend",
    "humidity_trend",
]

model_data = model_data.merge(
    weather_by_lap[weather_features],
    left_on="lap_number",
    right_index=True,
    how="left",
    validate="many_to_one",
)

# ---------------------------------------------------------
# 3. Define shared features for a fair comparison
# ---------------------------------------------------------
numeric_features = [
    "lap_number",
    "tyre_age",
    "rainfall",
    "humidity",
    "air_temperature",
    "track_temperature",
    "wind_speed",
    "traffic_ahead",
    "traffic_behind",
    "position_change",
    "rainfall_trend",
    "track_temp_trend",
    "air_temp_trend",
    "humidity_trend",
]

categorical_features = [
    "compound",
    "driver_number",
]

numeric_features = [
    col for col in numeric_features if col in model_data.columns
]

feature_columns = numeric_features + categorical_features

# Chronological split, matching Cell 24
split_lap = 35

train_data = model_data.loc[
    model_data["lap_number"] <= split_lap
].copy()

test_data = model_data.loc[
    model_data["lap_number"] > split_lap
].copy()

X_train = train_data[feature_columns]
y_train = train_data["lap_duration"]

X_test = test_data[feature_columns]
y_test = test_data["lap_duration"]

# ---------------------------------------------------------
# 4. Shared preprocessing
# ---------------------------------------------------------
def make_preprocessor(scale_numeric=False):
    numeric_steps = [
        ("imputer", SimpleImputer(strategy="median")),
    ]

    if scale_numeric:
        numeric_steps.append(("scaler", StandardScaler()))

    numeric_pipeline = Pipeline(numeric_steps)

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
    ])

    return ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
    ])

# ---------------------------------------------------------
# 5. Define models
# ---------------------------------------------------------
models = {
    "Linear Regression": Pipeline([
        ("preprocessor", make_preprocessor(scale_numeric=True)),
        ("model", LinearRegression()),
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", make_preprocessor()),
        ("model", RandomForestRegressor(
            n_estimators=300,
            min_samples_leaf=5,
            random_state=42,
            n_jobs=-1,
        )),
    ]),

    "XGBoost": Pipeline([
        ("preprocessor", make_preprocessor()),
        ("model", XGBRegressor(
            n_estimators=300,
            max_depth=4,
            learning_rate=0.03,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="reg:squarederror",
            random_state=42,
            n_jobs=-1,
        )),
    ]),
}

# ---------------------------------------------------------
# 6. Train and evaluate on exactly the same rows
# ---------------------------------------------------------
results = []
predictions_by_model = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    predictions_by_model[name] = predictions

    results.append({
        "model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2": r2_score(y_test, predictions),
    })

results_df = (
    pd.DataFrame(results)
    .sort_values("MAE")
    .reset_index(drop=True)
    .round(3)
)

print("MODEL COMPARISON: CHRONOLOGICAL TEST")
print("=" * 60)
print("Training laps: 1 to", split_lap)
print(
    "Test laps:",
    split_lap + 1,
    "to",
    int(test_data["lap_number"].max()),
)
print("Training rows:", len(train_data))
print("Test rows:", len(test_data))
display(results_df)

# ---------------------------------------------------------
# 7. Inspect best model by compound and lap
# ---------------------------------------------------------
best_model_name = results_df.iloc[0]["model"]
best_predictions = predictions_by_model[best_model_name]

diagnostics = test_data[
    ["lap_number", "compound", "lap_duration"]
].copy()

diagnostics["prediction"] = best_predictions
diagnostics["absolute_error"] = (
    diagnostics["lap_duration"] - diagnostics["prediction"]
).abs()

print(f"\nBEST MODEL BY MAE: {best_model_name}")

print("\nERROR BY COMPOUND")
display(
    diagnostics.groupby("compound")
    .agg(
        test_laps=("lap_duration", "size"),
        MAE=("absolute_error", "mean"),
        actual_median=("lap_duration", "median"),
        predicted_median=("prediction", "median"),
    )
    .round(3)
)

print("\nERROR BY LAP")
display(
    diagnostics.groupby("lap_number")
    .agg(
        actual_median=("lap_duration", "median"),
        predicted_median=("prediction", "median"),
        MAE=("absolute_error", "mean"),
    )
    .round(3)
)
 
```

```python

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Start from the existing dataset. Exclude pit-out laps from pace training.
pace_data = model_data.copy()

pace_data = pace_data.loc[
    ~pace_data["is_pit_out_lap"].fillna(False)
    & pace_data["lap_duration"].notna()
    & pace_data["compound"].notna()
].copy()

pace_data = pace_data.sort_values(
    ["driver_number", "lap_number"]
).reset_index(drop=True)

# All rolling features are shifted first, so the current lap's pace
# is never used to predict that same lap.
pace_data["previous_lap_time"] = (
    pace_data.groupby("driver_number")["lap_duration"].shift(1)
)

pace_data["previous_tyre_age"] = (
    pace_data.groupby("driver_number")["tyre_age"].shift(1)
)

pace_data["previous_rainfall"] = (
    pace_data.groupby("driver_number")["rainfall"].shift(1)
)

pace_data["previous_track_temperature"] = (
    pace_data.groupby("driver_number")["track_temperature"].shift(1)
)

pace_data["previous_humidity"] = (
    pace_data.groupby("driver_number")["humidity"].shift(1)
)

# Keep only rows with an available prior lap time.
pace_data = pace_data.loc[
    pace_data["previous_lap_time"].notna()
].copy()

numeric_features = [
    "lap_number",
    "tyre_age",
    "previous_tyre_age",
    "previous_lap_time",
    "previous_rainfall",
    "previous_track_temperature",
    "previous_humidity",
    "rainfall",
    "track_temperature",
    "humidity",
    "fuel_effect",
    "traffic_ahead",
    "traffic_behind",
    "tyre_degradation_effect",
]

numeric_features = [
    col for col in numeric_features if col in pace_data.columns
]

categorical_features = ["compound", "driver_number"]

# Chronological split, matching our previous evaluation.
train_mask = pace_data["lap_number"].between(2, 35)
test_mask = pace_data["lap_number"].between(36, 52)

train = pace_data.loc[train_mask].copy()
test = pace_data.loc[test_mask].copy()

X_train = train[numeric_features + categorical_features]
y_train = train["lap_duration"]

X_test = test[numeric_features + categorical_features]
y_test = test["lap_duration"]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler()),
            ]),
            numeric_features,
        ),
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore")),
            ]),
            categorical_features,
        ),
    ]
)

pace_model_ridge = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Ridge(alpha=10.0)),
])

pace_model_ridge.fit(X_train, y_train)
predictions = pace_model_ridge.predict(X_test)

# Baseline: median training lap time for the compound.
compound_medians = train.groupby("compound")["lap_duration"].median()
baseline_predictions = test["compound"].map(compound_medians)

# Use the same rows for both evaluations.
valid = baseline_predictions.notna().to_numpy()

def evaluate_predictions(y_true, y_pred, label):
    return {
        "model": label,
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred),
    }

results = pd.DataFrame([
    evaluate_predictions(y_test.to_numpy()[valid], predictions[valid],
                         "Lagged-feature Ridge"),
    evaluate_predictions(y_test.to_numpy()[valid],
                         baseline_predictions.to_numpy()[valid],
                         "Compound median baseline"),
])

print("Training rows:", len(train))
print("Test rows:", len(test))
print("Numeric features:", numeric_features)
display(results.round(3))
```

```python

# Cell 40: Diagnose pace by observed weather regime and tyre compound.
# This is analysis only. It does not train or modify any model.

import numpy as np
import pandas as pd

regime_data = model_data.copy()

# Exclude pit-out laps because they contain substantial pit-exit time.
regime_data = regime_data.loc[
    ~regime_data["is_pit_out_lap"].fillna(False)
    & regime_data["lap_duration"].notna()
    & regime_data["compound"].notna()
].copy()

# Use observed conditions to classify historical rows.
# These are retrospective diagnostic labels, not future inputs.
def classify_weather_regime(row):
    rainfall = row.get("rainfall", np.nan)
    grip = row.get("grip_condition", None)

    if pd.isna(rainfall):
        return "UNKNOWN"

    if rainfall >= 0.5:
        return "WET"

    if isinstance(grip, str) and grip.upper() == "LOW":
        return "DRYING_OR_LOW_GRIP"

    return "DRY_OR_NORMAL_GRIP"

regime_data["weather_regime"] = regime_data.apply(
    classify_weather_regime,
    axis=1
)

# Show how each compound performs under each observed regime.
regime_summary = (
    regime_data.groupby(
        ["weather_regime", "compound"],
        dropna=False
    )
    .agg(
        rows=("lap_duration", "size"),
        drivers=("driver_number", "nunique"),
        median_lap_time=("lap_duration", "median"),
        mean_lap_time=("lap_duration", "mean"),
        median_tyre_age=("tyre_age", "median"),
    )
    .reset_index()
    .sort_values(["weather_regime", "compound"])
)

display(regime_summary.round(3))

print("\nREGIME COVERAGE")
display(
    regime_data.groupby("weather_regime")
    .agg(
        rows=("lap_duration", "size"),
        drivers=("driver_number", "nunique"),
        first_lap=("lap_number", "min"),
        last_lap=("lap_number", "max"),
    )
    .reset_index()
)
```

```python

# Cell 42: Compare a weather-aware Ridge model against a compound-median baseline.

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# --------------------------------------------------
# 1. Prepare data
# --------------------------------------------------
pace_data = model_data.copy()

pace_data = pace_data.loc[
    (~pace_data["is_pit_out_lap"].fillna(False).astype(bool))
    & pace_data["lap_duration"].notna()
    & pace_data["compound"].notna()
    & pace_data["lap_number"].notna()
].copy()

# Keep a chronological split, matching our earlier experiments.
train_data = pace_data.loc[pace_data["lap_number"] <= 35].copy()
test_data = pace_data.loc[pace_data["lap_number"] >= 36].copy()

# Features available at the decision point.
# Trend features must already be lagged in model_data.
numeric_features = [
    "lap_number",
    "tyre_age",
    "fuel_effect",
    "traffic_ahead",
    "traffic_behind",
    "position_change",
    "rainfall",
    "humidity",
    "air_temperature",
    "track_temperature",
    "wind_speed",
    "rainfall_trend",
    "track_temp_trend",
    "air_temp_trend",
    "humidity_trend",
]

categorical_features = [
    "compound",
    "driver_number",
]

numeric_features = [
    c for c in numeric_features
    if c in pace_data.columns
]
categorical_features = [
    c for c in categorical_features
    if c in pace_data.columns
]

feature_columns = numeric_features + categorical_features

X_train = train_data[feature_columns].copy()
y_train = train_data["lap_duration"].copy()

X_test = test_data[feature_columns].copy()
y_test = test_data["lap_duration"].copy()

# --------------------------------------------------
# 2. Build preprocessing and Ridge regression
# --------------------------------------------------
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features),
])

ridge_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", Ridge(alpha=10.0)),
])

ridge_model.fit(X_train, y_train)
ridge_predictions = ridge_model.predict(X_test)

# --------------------------------------------------
# 3. Baseline: median pace by compound from training
# --------------------------------------------------
compound_medians = train_data.groupby("compound")[
    "lap_duration"
].median()

global_median = train_data["lap_duration"].median()

baseline_predictions = (
    test_data["compound"]
    .map(compound_medians)
    .fillna(global_median)
    .to_numpy()
)

# --------------------------------------------------
# 4. Evaluate both models on identical test rows
# --------------------------------------------------
evaluation = test_data[
    ["driver_number", "lap_number", "compound", "lap_duration"]
].copy().reset_index(drop=True)

evaluation["race_phase"] = pd.cut(
    evaluation["lap_number"],
    bins=[35, 39, 42, 46, 52],
    labels=[
        "WET_TO_DRY",
        "EARLY_DRY",
        "MID_LATE_RACE",
        "FINAL_LAPS",
    ],
    include_lowest=True,
)

evaluation["actual"] = evaluation["lap_duration"]
evaluation["ridge_prediction"] = ridge_predictions
evaluation["baseline_prediction"] = baseline_predictions

def calculate_metrics(frame, prediction_column):
    actual = frame["actual"]
    predicted = frame[prediction_column]
    return pd.Series({
        "rows": len(frame),
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "Bias": (predicted - actual).mean(),
        "R2": r2_score(actual, predicted),
    })

overall_results = pd.DataFrame({
    "Ridge": calculate_metrics(
        evaluation, "ridge_prediction"
    ),
    "Compound median": calculate_metrics(
        evaluation, "baseline_prediction"
    ),
}).T

print("OVERALL TEST RESULTS")
display(overall_results.round(3))

# --------------------------------------------------
# 5. Performance by race phase
# --------------------------------------------------
phase_results = []

for phase, group in evaluation.groupby(
    "race_phase", observed=True
):
    for model_name, prediction_column in [
        ("Ridge", "ridge_prediction"),
        ("Compound median", "baseline_prediction"),
    ]:
        result = calculate_metrics(group, prediction_column)
        phase_results.append({
            "race_phase": phase,
            "model": model_name,
            **result.to_dict(),
        })

phase_results = pd.DataFrame(phase_results)

print("\nRESULTS BY RACE PHASE")
display(
    phase_results
    .sort_values(["race_phase", "model"])
    .round(3)
)

# --------------------------------------------------
# 6. Performance by tyre compound
# --------------------------------------------------
compound_results = []

for compound, group in evaluation.groupby("compound"):
    for model_name, prediction_column in [
        ("Ridge", "ridge_prediction"),
        ("Compound median", "baseline_prediction"),
    ]:
        result = calculate_metrics(group, prediction_column)
        compound_results.append({
            "compound": compound,
            "model": model_name,
            **result.to_dict(),
        })

compound_results = pd.DataFrame(compound_results)

print("\nRESULTS BY TYRE COMPOUND")
display(
    compound_results
    .sort_values(["compound", "model"])
    .round(3)
)

# --------------------------------------------------
# 7. Inspect errors over time
# --------------------------------------------------
evaluation["ridge_error"] = (
    evaluation["ridge_prediction"] - evaluation["actual"]
)
evaluation["baseline_error"] = (
    evaluation["baseline_prediction"] - evaluation["actual"]
)

lap_errors = (
    evaluation.groupby("lap_number")
    .agg(
        rows=("actual", "size"),
        actual_median=("actual", "median"),
        ridge_median=("ridge_prediction", "median"),
        ridge_MAE=("ridge_error", lambda s: s.abs().mean()),
        ridge_bias=("ridge_error", "mean"),
        baseline_MAE=("baseline_error", lambda s: s.abs().mean()),
    )
    .reset_index()
)

print("\nERRORS BY LAP")
display(lap_errors.round(3))

print("\nTraining rows:", len(train_data))
print("Test rows:", len(test_data))
print("Features used:", feature_columns)
print("Training lap range:", train_data["lap_number"].min(),
      "to", train_data["lap_number"].max())
print("Test lap range:", test_data["lap_number"].min(),
      "to", test_data["lap_number"].max())
```

```python

# CELL 56 — PITWALL STRATEGY SIMULATOR
# Compare candidate pit timings over the remaining race.

from pathlib import Path
import numpy as np
import pandas as pd

# ------------------------------------------
# 1. Race decision state
# ------------------------------------------

DATA_DIR = Path("../data/processed/9558")

DRIVER = 63
CURRENT_LAP = 26
TOTAL_LAPS = 52
CURRENT_POSITION = 4

CURRENT_COMPOUND = "MEDIUM"
CURRENT_TYRE_AGE = 25

CANDIDATES = [
    {"strategy": "Stay out", "pit_lap": None,
     "compound": "MEDIUM"},
    {"strategy": "Pit lap 27", "pit_lap": 27,
     "compound": "INTERMEDIATE"},
    {"strategy": "Pit lap 28", "pit_lap": 28,
     "compound": "INTERMEDIATE"},
    {"strategy": "Pit lap 29", "pit_lap": 29,
     "compound": "INTERMEDIATE"},
]

# Replace this with your measured pit-lane loss if available.
# It must represent the additional time lost versus staying
# on track, not the entire pit-lap duration.
PIT_LOSS_SECONDS = 22.0

# ------------------------------------------
# 2. Load the frozen race-state assets
# ------------------------------------------

race = pd.read_parquet(DATA_DIR / "race_state.parquet")
tyres = pd.read_parquet(DATA_DIR / "tyre_features.parquet")

race = race[
    (race["session_key"] == 9558)
    & (race["lap_number"] > CURRENT_LAP)
    & (race["lap_number"] <= TOTAL_LAPS)
].copy()

# Use historical observed conditions as a provisional
# scenario, not as a claim about future weather.
driver_future = race[
    race["driver_number"] == DRIVER
].copy()

if driver_future.empty:
    raise ValueError("No future race-state rows for Driver 63.")

# ------------------------------------------
# 3. Build provisional lap-time scenarios
# ------------------------------------------

# The validated compound-median baseline has lower MAE
# than Ridge for INTERMEDIATE tyres in your evaluation.
# Use historical compound/tyre-age medians as a transparent
# fallback, rather than extrapolating Ridge to unsupported
# compound conditions.

historical = pd.read_parquet(DATA_DIR / "race_state.parquet")

# Exclude pit-out laps from ordinary running-lap baselines.
historical = historical[
    historical["lap_duration"].notna()
    & (~historical["is_pit_out_lap"].fillna(False))
].copy()

compound_age_baseline = (
    historical.groupby(
        ["compound", "tyre_age"], dropna=True
    )["lap_duration"]
    .median()
)

compound_baseline = (
    historical.groupby("compound")["lap_duration"]
    .median()
)

def baseline_lap_time(compound, tyre_age):
    key = (compound, tyre_age)

    if key in compound_age_baseline.index:
        return float(compound_age_baseline.loc[key])

    if compound in compound_baseline.index:
        return float(compound_baseline.loc[compound])

    raise ValueError(
        f"No historical baseline for compound {compound}"
    )

# ------------------------------------------
# 4. Simulate each candidate
# ------------------------------------------

results = []

for candidate in CANDIDATES:
    pit_lap = candidate["pit_lap"]
    compound_after_stop = candidate["compound"]

    projected_time = 0.0
    tyre_age = CURRENT_TYRE_AGE
    current_compound = CURRENT_COMPOUND
    pit_loss_added = False
    simulated_laps = []

    for lap in range(CURRENT_LAP + 1, TOTAL_LAPS + 1):

        # A pit on lap N means the driver runs that lap on
        # the old tyres, then changes compound for lap N+1.
        if pit_lap is not None and lap > pit_lap:
            current_compound = compound_after_stop
            if not pit_loss_added:
                projected_time += PIT_LOSS_SECONDS
                pit_loss_added = True
                tyre_age = 0

        tyre_age += 1

        lap_time = baseline_lap_time(
            current_compound, tyre_age
        )

        projected_time += lap_time

        simulated_laps.append({
            "lap": lap,
            "compound": current_compound,
            "tyre_age": tyre_age,
            "baseline_lap_time": lap_time,
        })

    results.append({
        "strategy": candidate["strategy"],
        "pit_lap": pit_lap,
        "planned_compound": compound_after_stop,
        "projected_remaining_time_s": projected_time,
        "pit_loss_s": (
            PIT_LOSS_SECONDS if pit_lap is not None else 0.0
        ),
        "simulated_laps": simulated_laps,
    })

# ------------------------------------------
# 5. Compare strategies
# ------------------------------------------

comparison = pd.DataFrame([
    {k: v for k, v in row.items()
     if k != "simulated_laps"}
    for row in results
])

comparison["difference_vs_best_s"] = (
    comparison["projected_remaining_time_s"]
    - comparison["projected_remaining_time_s"].min()
)

comparison = comparison.sort_values(
    "projected_remaining_time_s"
).reset_index(drop=True)

display(comparison)

print(
    "IMPORTANT: These are provisional compound/age baseline "
    "scenarios, not calibrated forecasts or win probabilities. "
    "The fixed pit loss and historical weather assumptions "
    "must be replaced with the existing strategy inputs before "
    "using this ranking as a race recommendation."
)
```

```python

# Cell 59: PITWALL strategy simulator v1
# Purpose: compare candidate strategies using historical wet/low-grip pace.
# This is a scenario model, not yet a calibrated race-outcome predictor.

import numpy as np
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/processed/9558")

DRIVER = 63
SESSION = 9558
CURRENT_LAP = 26
TOTAL_LAPS = 52
CURRENT_POSITION = 4
CURRENT_COMPOUND = "MEDIUM"
CURRENT_TYRE_AGE = 25

# Explicit assumptions: test several pit-loss values rather than hiding uncertainty.
PIT_LOSS_SCENARIOS = [18.0, 22.0, 26.0]

CANDIDATES = [
    {"strategy": "Stay out", "pit_lap": None, "compound": "MEDIUM"},
    {"strategy": "Pit lap 27", "pit_lap": 27, "compound": "INTERMEDIATE"},
    {"strategy": "Pit lap 28", "pit_lap": 28, "compound": "INTERMEDIATE"},
    {"strategy": "Pit lap 29", "pit_lap": 29, "compound": "INTERMEDIATE"},
]

race = pd.read_parquet(DATA_DIR / "race_state.parquet")

# Use historical observations from this session in wet, low-grip conditions.
hist = race[
    (race["session_key"] == SESSION)
    & race["lap_duration"].notna()
    & (~race["is_pit_out_lap"].fillna(False))
    & (race["rainfall"] == 1.0)
    & (race["grip_condition"] == "LOW")
].copy()

hist["compound"] = hist["compound"].str.upper()
hist = hist[hist["tyre_age"].notna() & (hist["tyre_age"] >= 0)]

# Use tyre-age bands to avoid pretending every exact age has enough observations.
AGE_BINS = [-1, 5, 10, 15, 20, 30, 100]
AGE_LABELS = ["0-5", "6-10", "11-15", "16-20", "21-30", "31+"]

hist["age_band"] = pd.cut(
    hist["tyre_age"],
    bins=AGE_BINS,
    labels=AGE_LABELS
)

pace_table = (
    hist.groupby(["compound", "age_band"], observed=True)["lap_duration"]
    .agg(["median", "count"])
    .reset_index()
)

# Anchor the current compound's age band to the actual decision-point lap time.
current_row = race[
    (race["session_key"] == SESSION)
    & (race["driver_number"] == DRIVER)
    & (race["lap_number"] == CURRENT_LAP)
]

if current_row.empty:
    raise ValueError("Could not find driver #63 at the decision lap.")

current_row = current_row.iloc[0]
CURRENT_LAP_TIME = float(current_row["lap_duration"])

def age_band(age):
    for upper, label in zip([5, 10, 15, 20, 30, 100], AGE_LABELS):
        if age <= upper:
            return label
    return "31+"

def empirical_pace(compound, age):
    """Return the historical median for compound/age band.
    Fall back to the nearest available age band for that compound.
    """
    band = age_band(age)
    rows = pace_table[
        (pace_table["compound"] == compound)
        & (pace_table["age_band"] == band)
    ]

    if not rows.empty:
        row = rows.iloc[0]
        return float(row["median"]), int(row["count"]), band

    available = pace_table[
        (pace_table["compound"] == compound)
        & (pace_table["count"] >= 5)
    ].copy()

    if available.empty:
        raise ValueError(
            f"No usable historical pace observations for {compound}."
        )

    # Find the closest available tyre-age band.
    band_midpoint = {
        "0-5": 2.5, "6-10": 8, "11-15": 13,
        "16-20": 18, "21-30": 25.5, "31+": 35
    }
    available["distance"] = available["age_band"].astype(str).map(
        lambda x: abs(band_midpoint[x] - age)
    )
    row = available.sort_values("distance").iloc[0]
    return float(row["median"]), int(row["count"]), str(row["age_band"])

# Align the Medium historical pace curve to the actual observed lap at lap 26.
medium_hist, _, _ = empirical_pace("MEDIUM", CURRENT_TYRE_AGE)
medium_anchor_adjustment = CURRENT_LAP_TIME - medium_hist

def projected_lap_time(compound, age):
    pace, count, source_band = empirical_pace(compound, age)

    # Preserve the target driver's current pace level for the Medium curve.
    # Intermediate pace remains a historical estimate, not an XGBoost prediction.
    if compound == "MEDIUM":
        pace += medium_anchor_adjustment

    return pace, count, source_band

def simulate(candidate, pit_loss):
    pit_lap = candidate["pit_lap"]
    new_compound = candidate["compound"]

    compound = CURRENT_COMPOUND
    tyre_age = CURRENT_TYRE_AGE
    total_time = 0.0
    pit_loss_added = False
    lap_records = []

    for lap in range(CURRENT_LAP + 1, TOTAL_LAPS + 1):

        # The driver completes the pit-in lap on the current tyres.
        if pit_lap is not None and lap == pit_lap:
            tyre_age += 1
            lap_time, n, band = projected_lap_time(compound, tyre_age)
            total_time += lap_time + pit_loss
            pit_loss_added = True

            lap_records.append({
                "lap": lap,
                "compound": compound,
                "tyre_age": tyre_age,
                "estimated_lap_time": lap_time,
                "pit_loss": pit_loss,
                "pace_sample_count": n,
                "pace_age_band": band
            })

            # New tyres are fitted after the pit-in lap.
            compound = new_compound
            tyre_age = 0
            continue

        tyre_age += 1
        lap_time, n, band = projected_lap_time(compound, tyre_age)
        total_time += lap_time

        lap_records.append({
            "lap": lap,
            "compound": compound,
            "tyre_age": tyre_age,
            "estimated_lap_time": lap_time,
            "pit_loss": 0.0,
            "pace_sample_count": n,
            "pace_age_band": band
        })

    return total_time, lap_records

results = []
details = {}

for pit_loss in PIT_LOSS_SCENARIOS:
    scenario_results = []

    for candidate in CANDIDATES:
        total, laps = simulate(candidate, pit_loss)
        scenario_results.append({
            "pit_loss_assumption_s": pit_loss,
            "strategy": candidate["strategy"],
            "pit_lap": candidate["pit_lap"],
            "new_compound": candidate["compound"],
            "projected_remaining_time_s": round(total, 2),
        })
        details[(pit_loss, candidate["strategy"])] = pd.DataFrame(laps)

    best_time = min(x["projected_remaining_time_s"] for x in scenario_results)

    for result in scenario_results:
        result["gap_to_best_s"] = round(
            result["projected_remaining_time_s"] - best_time, 2
        )
        results.append(result)

comparison = pd.DataFrame(results)

display(
    comparison.sort_values(
        ["pit_loss_assumption_s", "projected_remaining_time_s"]
    ).reset_index(drop=True)
)

print(
    f"Decision: after lap {CURRENT_LAP}, driver #{DRIVER}, "
    f"P{CURRENT_POSITION}, current lap time {CURRENT_LAP_TIME:.3f}s."
)
print(
    "Important: these are conditional scenario estimates. "
    "They do not yet simulate rival strategies or predict a finishing position."
)
```

```python

# Cell 60: Full-field strategy comparison
# Baseline assumption: rivals continue on their current compounds.
# Target driver #63 is evaluated under each candidate strategy.

PIT_LOSS = 22.0  # Assumption in seconds; change for sensitivity testing.

field = race[
    (race["session_key"] == SESSION)
    & (race["lap_number"] == CURRENT_LAP)
].copy()

field = field[
    field["position"].notna()
    & field["compound"].notna()
    & field["tyre_age"].notna()
].copy()

field["compound"] = field["compound"].str.upper()

# Elapsed time through the decision lap, using observed lap times.
elapsed = (
    race[
        (race["session_key"] == SESSION)
        & (race["lap_number"] <= CURRENT_LAP)
        & race["lap_duration"].notna()
    ]
    .groupby("driver_number")["lap_duration"]
    .sum()
    .rename("elapsed_time_s")
)

field = field.merge(elapsed, on="driver_number", how="left")
field = field.dropna(subset=["elapsed_time_s"])

def project_driver_remaining(compound, tyre_age, driver_number,
                             strategy_compound=None, pit_lap=None):
    total = 0.0
    age = int(tyre_age)
    active_compound = str(compound).upper()

    for lap in range(CURRENT_LAP + 1, TOTAL_LAPS + 1):
        if pit_lap is not None and lap == pit_lap:
            # Complete pit-in lap on the old tyres, then add pit loss.
            age += 1
            lap_time, _, _ = projected_lap_time(active_compound, age)
            total += lap_time + PIT_LOSS

            active_compound = strategy_compound
            age = 0
            continue

        age += 1
        lap_time, _, _ = projected_lap_time(active_compound, age)
        total += lap_time

    return total

# Project every rival under the baseline of no further pit stops.
baseline_remaining = {}

for _, row in field.iterrows():
    number = int(row["driver_number"])
    baseline_remaining[number] = project_driver_remaining(
        compound=row["compound"],
        tyre_age=row["tyre_age"],
        driver_number=number
    )

# Replace driver #63's baseline with each candidate strategy.
strategy_rows = []

for candidate in CANDIDATES:
    target_number = DRIVER
    target_row = field[field["driver_number"] == target_number]

    if target_row.empty:
        raise ValueError("Driver #63 is missing from the lap-26 field.")

    target_row = target_row.iloc[0]

    target_remaining = project_driver_remaining(
        compound=target_row["compound"],
        tyre_age=target_row["tyre_age"],
        driver_number=target_number,
        strategy_compound=candidate["compound"],
        pit_lap=candidate["pit_lap"]
    )

    projected = []

    for _, row in field.iterrows():
        number = int(row["driver_number"])
        remaining = (
            target_remaining
            if number == target_number
            else baseline_remaining[number]
        )

        projected.append({
            "driver_number": number,
            "current_position": int(row["position"]),
            "projected_total_time_s": (
                float(row["elapsed_time_s"]) + remaining
            )
        })

    ranked = pd.DataFrame(projected).sort_values(
        "projected_total_time_s"
    ).reset_index(drop=True)

    ranked["projected_position"] = np.arange(1, len(ranked) + 1)

    target_result = ranked[ranked["driver_number"] == DRIVER].iloc[0]

    strategy_rows.append({
        "strategy": candidate["strategy"],
        "pit_lap": candidate["pit_lap"],
        "projected_position": int(target_result["projected_position"]),
        "current_position": CURRENT_POSITION,
        "position_change": (
            CURRENT_POSITION - int(target_result["projected_position"])
        ),
        "projected_total_time_s": round(
            float(target_result["projected_total_time_s"]), 2
        ),
        "gap_to_projected_leader_s": round(
            float(target_result["projected_total_time_s"])
            - float(ranked.iloc[0]["projected_total_time_s"]),
            2
        )
    })

    print(f"\n{candidate['strategy']} | Projected field order:")
    display(
        ranked[["projected_position", "driver_number",
                "current_position", "projected_total_time_s"]]
    )

field_comparison = pd.DataFrame(strategy_rows).sort_values(
    ["projected_position", "projected_total_time_s"]
)

print("\nPITWALL: strategy impact on projected finishing position")
display(field_comparison.reset_index(drop=True))
```

```python

# Cell 61: Driver-specific pace anchoring
# Reuse the existing variables from Cells 59-60.

def project_driver_remaining_v2(
    driver_number,
    compound,
    tyre_age,
    strategy_compound=None,
    pit_lap=None
):
    # Get this driver's own observed lap time at the decision point.
    observed = race[
        (race["session_key"] == SESSION)
        & (race["driver_number"] == driver_number)
        & (race["lap_number"] == CURRENT_LAP)
    ]

    if observed.empty:
        raise ValueError(f"No decision-lap observation for driver {driver_number}")

    observed_time = float(observed.iloc[0]["lap_duration"])
    compound = str(compound).upper()

    # Anchor this driver's starting compound curve to their observed pace.
    hist_pace, _, _ = empirical_pace(compound, int(tyre_age))
    driver_adjustment = observed_time - hist_pace

    total = 0.0
    age = int(tyre_age)
    active_compound = compound

    for lap in range(CURRENT_LAP + 1, TOTAL_LAPS + 1):
        if pit_lap is not None and lap == pit_lap:
            age += 1
            pace, _, _ = empirical_pace(active_compound, age)
            total += pace + driver_adjustment + PIT_LOSS

            active_compound = str(strategy_compound).upper()
            age = 0

            # For the new compound, use the historical pace curve without
            # pretending that the old-compound adjustment is transferable.
            driver_adjustment = 0.0
            continue

        age += 1
        pace, _, _ = empirical_pace(active_compound, age)
        total += pace + driver_adjustment

    return total


# Recalculate each rival's baseline with their own pace anchor.
baseline_remaining_v2 = {}

for _, row in field.iterrows():
    number = int(row["driver_number"])
    baseline_remaining_v2[number] = project_driver_remaining_v2(
        driver_number=number,
        compound=row["compound"],
        tyre_age=row["tyre_age"]
    )

# Re-evaluate each target strategy against the driver-specific field baseline.
results_v2 = []

for candidate in CANDIDATES:
    target_row = field[field["driver_number"] == DRIVER].iloc[0]

    target_remaining = project_driver_remaining_v2(
        driver_number=DRIVER,
        compound=target_row["compound"],
        tyre_age=target_row["tyre_age"],
        strategy_compound=candidate["compound"],
        pit_lap=candidate["pit_lap"]
    )

    projected = []

    for _, row in field.iterrows():
        number = int(row["driver_number"])
        remaining = (
            target_remaining
            if number == DRIVER
            else baseline_remaining_v2[number]
        )

        projected.append({
            "driver_number": number,
            "current_position": int(row["position"]),
            "projected_total_time_s": float(row["elapsed_time_s"]) + remaining
        })

    ranked = pd.DataFrame(projected).sort_values(
        "projected_total_time_s"
    ).reset_index(drop=True)

    ranked["projected_position"] = np.arange(1, len(ranked) + 1)
    target_result = ranked[ranked["driver_number"] == DRIVER].iloc[0]

    results_v2.append({
        "strategy": candidate["strategy"],
        "pit_lap": candidate["pit_lap"],
        "projected_position": int(target_result["projected_position"]),
        "position_change": CURRENT_POSITION - int(target_result["projected_position"]),
        "projected_total_time_s": round(float(target_result["projected_total_time_s"]), 2),
        "gap_to_projected_leader_s": round(
            float(target_result["projected_total_time_s"])
            - float(ranked.iloc[0]["projected_total_time_s"]), 2
        )
    })

field_comparison_v2 = pd.DataFrame(results_v2).sort_values(
    ["projected_position", "projected_total_time_s"]
)

display(field_comparison_v2.reset_index(drop=True))
```

```python

# Cell 62: Consistent driver pace + tyre-relative strategy simulation
# All drivers are projected over the same remaining laps.
# Driver pace adjustment stays constant across compound changes.

PIT_LOSS = 22.0  # Assumption in seconds; test sensitivity later.

# Keep only drivers with a decision-lap observation and usable elapsed time.
field62 = field.copy()
field62["driver_number"] = field62["driver_number"].astype(int)
field62["compound"] = field62["compound"].str.upper()

# Historical median pace for each compound and tyre-age band.
# Require at least 5 observations; fallback to the nearest supported age band.
def robust_empirical_pace(compound, age):
    band = age_band(age)

    rows = pace_table[
        (pace_table["compound"] == compound)
        & (pace_table["age_band"].astype(str) == band)
        & (pace_table["count"] >= 5)
    ]

    if not rows.empty:
        row = rows.iloc[0]
        return float(row["median"]), int(row["count"]), str(row["age_band"])

    available = pace_table[
        (pace_table["compound"] == compound)
        & (pace_table["count"] >= 5)
    ].copy()

    if available.empty:
        raise ValueError(
            f"No supported historical pace data for compound={compound}"
        )

    band_midpoint = {
        "0-5": 2.5, "6-10": 8, "11-15": 13,
        "16-20": 18, "21-30": 25.5, "31+": 35
    }

    available["distance"] = available["age_band"].astype(str).map(
        lambda x: abs(band_midpoint[x] - age)
    )

    row = available.sort_values("distance").iloc[0]
    return float(row["median"]), int(row["count"]), str(row["age_band"])


def simulate_driver62(
    driver_number,
    starting_compound,
    starting_age,
    observed_lap_time,
    pit_lap=None,
    new_compound=None,
    pit_loss=PIT_LOSS
):
    starting_compound = str(starting_compound).upper()

    # Driver-specific pace level, estimated at the decision point.
    reference_pace, _, _ = robust_empirical_pace(
        starting_compound, int(starting_age)
    )
    driver_offset = float(observed_lap_time) - reference_pace

    total_lap_time = 0.0
    total_pit_loss = 0.0
    age = int(starting_age)
    compound = starting_compound
    rows = []

    for lap in range(CURRENT_LAP + 1, TOTAL_LAPS + 1):

        # Pit-in lap is completed on the old tyres.
        if pit_lap is not None and lap == pit_lap:
            age += 1
            pace, n, used_band = robust_empirical_pace(compound, age)
            projected_lap = pace + driver_offset

            total_lap_time += projected_lap
            total_pit_loss += pit_loss

            rows.append({
                "driver_number": driver_number,
                "lap": lap,
                "compound": compound,
                "tyre_age": age,
                "projected_lap_time_s": projected_lap,
                "pit_loss_s": pit_loss,
                "pace_sample_count": n,
                "pace_age_band": used_band
            })

            compound = str(new_compound).upper()
            age = 0
            continue

        age += 1
        pace, n, used_band = robust_empirical_pace(compound, age)
        projected_lap = pace + driver_offset

        total_lap_time += projected_lap

        rows.append({
            "driver_number": driver_number,
            "lap": lap,
            "compound": compound,
            "tyre_age": age,
            "projected_lap_time_s": projected_lap,
            "pit_loss_s": 0.0,
            "pace_sample_count": n,
            "pace_age_band": used_band
        })

    return total_lap_time, total_pit_loss, pd.DataFrame(rows)


# Build one consistent baseline for every driver.
driver_models62 = {}

for _, row in field62.iterrows():
    number = int(row["driver_number"])

    observed = race[
        (race["session_key"] == SESSION)
        & (race["driver_number"] == number)
        & (race["lap_number"] == CURRENT_LAP)
    ]

    if observed.empty or pd.isna(observed.iloc[0]["lap_duration"]):
        continue

    row_time = float(observed.iloc[0]["lap_duration"])

    remaining, pit_cost, lap_detail = simulate_driver62(
        driver_number=number,
        starting_compound=row["compound"],
        starting_age=int(row["tyre_age"]),
        observed_lap_time=row_time
    )

    driver_models62[number] = {
        "remaining_time_s": remaining + pit_cost,
        "elapsed_time_s": float(row["elapsed_time_s"]),
        "current_position": int(row["position"]),
        "lap_detail": lap_detail
    }


# Evaluate #63's strategies against the same baseline for all rivals.
strategy_results62 = []
rankings62 = {}

for candidate in CANDIDATES:
    target = field62[field62["driver_number"] == DRIVER]

    if target.empty or DRIVER not in driver_models62:
        raise ValueError("Driver #63 is missing from the usable field.")

    target = target.iloc[0]

    target_remaining, target_pit_loss, target_laps = simulate_driver62(
        driver_number=DRIVER,
        starting_compound=target["compound"],
        starting_age=int(target["tyre_age"]),
        observed_lap_time=float(
            race[
                (race["session_key"] == SESSION)
                & (race["driver_number"] == DRIVER)
                & (race["lap_number"] == CURRENT_LAP)
            ].iloc[0]["lap_duration"]
        ),
        pit_lap=candidate["pit_lap"],
        new_compound=candidate["compound"],
        pit_loss=PIT_LOSS
    )

    projected_rows = []

    for number, model in driver_models62.items():
        remaining = (
            target_remaining + target_pit_loss
            if number == DRIVER
            else model["remaining_time_s"]
        )

        projected_rows.append({
            "driver_number": number,
            "current_position": model["current_position"],
            "elapsed_time_s": model["elapsed_time_s"],
            "remaining_time_s": remaining,
            "projected_total_time_s": (
                model["elapsed_time_s"] + remaining
            )
        })

    ranking = pd.DataFrame(projected_rows).sort_values(
        "projected_total_time_s"
    ).reset_index(drop=True)

    ranking["projected_position"] = np.arange(1, len(ranking) + 1)
    rankings62[candidate["strategy"]] = ranking

    target_result = ranking[
        ranking["driver_number"] == DRIVER
    ].iloc[0]

    leader_time = float(ranking.iloc[0]["projected_total_time_s"])

    strategy_results62.append({
        "strategy": candidate["strategy"],
        "pit_lap": candidate["pit_lap"],
        "projected_position": int(target_result["projected_position"]),
        "position_change": (
            CURRENT_POSITION - int(target_result["projected_position"])
        ),
        "projected_remaining_lap_time_s": round(target_remaining, 2),
        "pit_loss_s": round(target_pit_loss, 2),
        "projected_total_time_s": round(
            float(target_result["projected_total_time_s"]), 2
        ),
        "gap_to_projected_leader_s": round(
            float(target_result["projected_total_time_s"]) - leader_time, 2
        )
    })

comparison62 = pd.DataFrame(strategy_results62).sort_values(
    "projected_total_time_s"
).reset_index(drop=True)

display(comparison62)

# Consistency check: every rival uses the same baseline across strategies.
print(
    "Drivers included:", len(driver_models62),
    "| Laps simulated per strategy:", TOTAL_LAPS - CURRENT_LAP
)
print("Pit-loss assumption:", PIT_LOSS, "seconds")
print(
    "Reminder: this is a simplified scenario model. "
    "It assumes wet/low-grip pace persists and rivals do not pit."
)
```

```python
# PITWALL — Baseline vs weather-enhanced pace prediction
# Diagnostic only: does adding weather improve chronological test performance?

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

weather_test_data = race_state.loc[
    race_state["lap_duration"].notna()
    & race_state["compound"].notna()
    & ~race_state["is_pit_out_lap"].fillna(False).astype(bool)
].copy()

weather_test_data["compound"] = (
    weather_test_data["compound"].astype(str).str.upper()
)
weather_test_data["driver_number"] = (
    weather_test_data["driver_number"].astype(str)
)

baseline_numeric = ["lap_number", "tyre_age"]
weather_numeric = [
    "lap_number", "tyre_age",
    "rainfall", "track_temperature",
    "air_temperature", "humidity", "wind_speed"
]
categorical = ["compound", "driver_number"]

train = weather_test_data.loc[
    weather_test_data["lap_number"] <= 35
].copy()

test = weather_test_data.loc[
    weather_test_data["lap_number"].between(36, 52)
].copy()

def evaluate_weather_model(numeric_features, model_name):
    preprocessor = ColumnTransformer([
        ("numeric", Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ]), numeric_features),
        ("categorical", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical)
    ])

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=300,
            min_samples_leaf=5,
            random_state=42,
            n_jobs=-1
        ))
    ])

    features = numeric_features + categorical
    model.fit(train[features], train["lap_duration"])
    predictions = model.predict(test[features])

    return {
        "model": model_name,
        "train_rows": len(train),
        "test_rows": len(test),
        "MAE_s": mean_absolute_error(test["lap_duration"], predictions),
        "RMSE_s": np.sqrt(
            mean_squared_error(test["lap_duration"], predictions)
        ),
        "R2": r2_score(test["lap_duration"], predictions)
    }

results_weather = pd.DataFrame([
    evaluate_weather_model(baseline_numeric, "Baseline"),
    evaluate_weather_model(weather_numeric, "Baseline + weather")
])

print("CHRONOLOGICAL VALIDATION: TRAIN LAPS 1–35, TEST LAPS 36–52")
display(results_weather.round(4))

print("Weather test-period coverage:")
display(
    test[weather_numeric[2:]]
    .isna()
    .sum()
    .to_frame("missing_test_values")
)
```

```python

from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor

# ============================================================
# CELL 63: TEMPORAL PACE-MODEL BACKTEST
# ============================================================

DATA_DIR = Path("../data/processed/9558")
SESSION = 9558

# ------------------------------------------------------------
# 1. Load existing processed datasets
# ------------------------------------------------------------

state = pd.read_parquet(DATA_DIR / "race_state.parquet")
fuel = pd.read_parquet(DATA_DIR / "fuel_features.parquet")
traffic = pd.read_parquet(DATA_DIR / "traffic_features.parquet")
tyres = pd.read_parquet(DATA_DIR / "tyre_features.parquet")

# Keep only the features required from each asset
fuel = fuel[
    ["driver_number", "lap_number", "fuel_effect"]
].drop_duplicates(["driver_number", "lap_number"])

traffic = traffic[
    [
        "driver_number",
        "lap_number",
        "traffic_ahead",
        "traffic_behind",
        "position_change",
    ]
].drop_duplicates(["driver_number", "lap_number"])

tyres = tyres[
    ["driver_number", "lap_number", "tyre_degradation_effect"]
].drop_duplicates(["driver_number", "lap_number"])

# ------------------------------------------------------------
# 2. Filter session and merge features
# ------------------------------------------------------------

df = state[state["session_key"] == SESSION].copy()

df = df.merge(
    fuel,
    on=["driver_number", "lap_number"],
    how="left",
    validate="many_to_one",
)

df = df.merge(
    traffic,
    on=["driver_number", "lap_number"],
    how="left",
    validate="many_to_one",
)

df = df.merge(
    tyres,
    on=["driver_number", "lap_number"],
    how="left",
    validate="many_to_one",
)

# ------------------------------------------------------------
# 3. Create lagged track-evolution feature
# ------------------------------------------------------------

df["lap_duration"] = pd.to_numeric(
    df["lap_duration"], errors="coerce"
)

field_median = (
    df.groupby("lap_number")["lap_duration"]
    .median()
    .sort_index()
)

previous_field_median = field_median.shift(1)

df["track_evolution"] = df["lap_number"].map(
    previous_field_median
)

first_available = previous_field_median.dropna()

if first_available.empty:
    raise ValueError(
        "Cannot calculate track evolution: no previous-lap median available."
    )

df["track_evolution"] = df["track_evolution"].fillna(
    float(first_available.iloc[0])
)

# ------------------------------------------------------------
# 4. Clean target and filter compounds
# ------------------------------------------------------------

df["compound"] = (
    df["compound"].astype("string").str.upper()
)

df = df[
    df["lap_duration"].notna()
    & np.isfinite(df["lap_duration"])
    & (df["lap_duration"] > 0)
    & (~df["is_pit_out_lap"].fillna(False).astype(bool))
    & df["compound"].isin(["MEDIUM", "SOFT"])
].copy()

df = df.reset_index(drop=True)

if df.empty:
    raise ValueError(
        "No valid MEDIUM/SOFT laps remain after filtering."
    )

print("Valid rows after cleaning:", len(df))
print("Invalid lap durations remaining:", (
    df["lap_duration"].isna()
    | ~np.isfinite(df["lap_duration"])
).sum())

# ------------------------------------------------------------
# 5. Build features
# ------------------------------------------------------------

numeric_features = [
    "tyre_age",
    "fuel_effect",
    "traffic_ahead",
    "traffic_behind",
    "position_change",
    "track_evolution",
    "rainfall",
    "humidity",
    "air_temperature",
    "wind_speed",
]

missing_features = [
    col for col in numeric_features if col not in df.columns
]

if missing_features:
    raise KeyError(f"Missing feature columns: {missing_features}")

numeric_X = df[numeric_features].apply(
    pd.to_numeric, errors="coerce"
)

driver_dummies = pd.get_dummies(
    df["driver_number"].astype(str),
    prefix="driver",
    dtype=int,
)

compound_dummies = pd.get_dummies(
    df["compound"],
    prefix="compound",
    dtype=int,
)

X = pd.concat(
    [
        numeric_X.reset_index(drop=True),
        driver_dummies.reset_index(drop=True),
        compound_dummies.reset_index(drop=True),
    ],
    axis=1,
)

y = df["lap_duration"].reset_index(drop=True)
laps = pd.to_numeric(
    df["lap_number"], errors="coerce"
).reset_index(drop=True)
compounds = df["compound"].reset_index(drop=True)

# Remove any rows with invalid lap numbers, keeping all arrays aligned
valid_lap_mask = laps.notna().to_numpy()

X = X.iloc[np.flatnonzero(valid_lap_mask)].reset_index(drop=True)
y = y.iloc[np.flatnonzero(valid_lap_mask)].reset_index(drop=True)
laps = laps.iloc[np.flatnonzero(valid_lap_mask)].reset_index(drop=True)
compounds = compounds.iloc[
    np.flatnonzero(valid_lap_mask)
].reset_index(drop=True)

# ------------------------------------------------------------
# 6. Temporal train/test split
#    Train on earlier laps, test on later laps
# ------------------------------------------------------------

split_lap = int(laps.quantile(0.70))

train_idx = np.flatnonzero((laps <= split_lap).to_numpy())
test_idx = np.flatnonzero((laps > split_lap).to_numpy())

if len(train_idx) == 0 or len(test_idx) == 0:
    raise ValueError(
        f"Empty temporal split. Split lap: {split_lap}"
    )

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

train_compounds = compounds.iloc[train_idx].reset_index(drop=True)
test_compounds = compounds.iloc[test_idx].reset_index(drop=True)

# ------------------------------------------------------------
# 7. Handle missing feature values using training data only
# ------------------------------------------------------------

X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

train_medians = X_train.median(numeric_only=True)

X_train = X_train.fillna(train_medians).fillna(0)
X_test = X_test.fillna(train_medians).fillna(0)

# Final target safety checks
assert np.isfinite(y_train.to_numpy(dtype=float)).all(), (
    "Invalid values remain in y_train."
)
assert np.isfinite(y_test.to_numpy(dtype=float)).all(), (
    "Invalid values remain in y_test."
)

print(f"\nTemporal split lap: {split_lap}")
print(f"Training rows: {len(X_train)}")
print(f"Testing rows: {len(X_test)}")
print(f"Training lap range: {laps.iloc[train_idx].min()}–"
      f"{laps.iloc[train_idx].max()}")
print(f"Testing lap range: {laps.iloc[test_idx].min()}–"
      f"{laps.iloc[test_idx].max()}")

# ------------------------------------------------------------
# 8. Train XGBoost pace model
# ------------------------------------------------------------

model = XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.04,
    subsample=0.85,
    colsample_bytree=0.85,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1,
)

model.fit(X_train, y_train)

pred = model.predict(X_test)

# ------------------------------------------------------------
# 9. Baseline: training-set median lap time by compound
# ------------------------------------------------------------

compound_medians = (
    pd.DataFrame({
        "compound": train_compounds,
        "lap_time": y_train.reset_index(drop=True),
    })
    .groupby("compound")["lap_time"]
    .median()
)

baseline_pred = test_compounds.map(
    compound_medians
).to_numpy(dtype=float)

# A test compound absent from training cannot use this baseline.
baseline_valid = np.isfinite(baseline_pred)

# ------------------------------------------------------------
# 10. Evaluate both approaches
# ------------------------------------------------------------

def metrics(name, actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    return {
        "model": name,
        "n": len(actual),
        "MAE_seconds": mean_absolute_error(actual, predicted),
        "RMSE_seconds": np.sqrt(
            mean_squared_error(actual, predicted)
        ),
        "Bias_seconds": float(np.mean(predicted - actual)),
        "R2": r2_score(actual, predicted),
    }

results = [
    metrics("Temporal XGBoost", y_test, pred)
]

if baseline_valid.any():
    results.append(
        metrics(
            "Training compound median",
            y_test.to_numpy()[baseline_valid],
            baseline_pred[baseline_valid],
        )
    )

results_df = pd.DataFrame(results)

print("\n========== TEMPORAL BACKTEST RESULTS ==========")
print(results_df.round(4).to_string(index=False))

# ------------------------------------------------------------
# 11. Inspect prediction errors
# ------------------------------------------------------------

predictions = pd.DataFrame({
    "lap_number": laps.iloc[test_idx].to_numpy(),
    "driver_number": df.loc[
        valid_lap_mask, "driver_number"
    ].to_numpy()[test_idx],
    "compound": test_compounds.to_numpy(),
    "actual_lap_time": y_test.to_numpy(),
    "predicted_lap_time": pred,
})

predictions["error_seconds"] = (
    predictions["predicted_lap_time"]
    - predictions["actual_lap_time"]
)

print("\n========== SAMPLE TEST PREDICTIONS ==========")
print(predictions.head(10).round(3).to_string(index=False))

print("\nMean absolute prediction error:",
      round(predictions["error_seconds"].abs().mean(), 3),
      "seconds")
```

```python

from pathlib import Path
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor

DATA_DIR = Path("../data/processed/9558")
SESSION = 9558

# 1. Load and merge existing assets
state = pd.read_parquet(DATA_DIR / "race_state.parquet")
fuel = pd.read_parquet(DATA_DIR / "fuel_features.parquet")
traffic = pd.read_parquet(DATA_DIR / "traffic_features.parquet")
tyres = pd.read_parquet(DATA_DIR / "tyre_features.parquet")

fuel = fuel[
    ["driver_number", "lap_number", "fuel_effect"]
].drop_duplicates(["driver_number", "lap_number"])

traffic = traffic[
    ["driver_number", "lap_number", "traffic_ahead",
     "traffic_behind", "position_change"]
].drop_duplicates(["driver_number", "lap_number"])

tyres = tyres[
    ["driver_number", "lap_number", "tyre_degradation_effect"]
].drop_duplicates(["driver_number", "lap_number"])

df = state[state["session_key"] == SESSION].copy()

for asset in [fuel, traffic, tyres]:
    df = df.merge(
        asset,
        on=["driver_number", "lap_number"],
        how="left",
        validate="many_to_one",
    )

# 2. Clean target, but keep ALL tyre compounds
df["lap_duration"] = pd.to_numeric(df["lap_duration"], errors="coerce")
df["compound"] = df["compound"].astype("string").str.upper()

df = df[
    df["lap_duration"].notna()
    & np.isfinite(df["lap_duration"])
    & (df["lap_duration"] > 0)
    & (~df["is_pit_out_lap"].fillna(False).astype(bool))
    & df["compound"].notna()
].copy()

df = df.reset_index(drop=True)

# 3. Build track evolution from the previous completed lap only
field_median = df.groupby("lap_number")["lap_duration"].median().sort_index()
previous_field_median = field_median.shift(1)
df["track_evolution"] = df["lap_number"].map(previous_field_median)

# 4. Create shared features for all models
numeric_features = [
    "tyre_age",
    "tyre_degradation_effect",
    "fuel_effect",
    "traffic_ahead",
    "traffic_behind",
    "position_change",
    "track_evolution",
    "rainfall",
    "humidity",
    "air_temperature",
    "wind_speed",
]

missing = [c for c in numeric_features if c not in df.columns]
if missing:
    raise KeyError(f"Missing expected features: {missing}")

X_num = df[numeric_features].apply(pd.to_numeric, errors="coerce")
X_num = X_num.replace([np.inf, -np.inf], np.nan)

# Include compound and driver categories
compound_dummies = pd.get_dummies(
    df["compound"], prefix="compound", dtype=int
)
driver_dummies = pd.get_dummies(
    df["driver_number"].astype(str), prefix="driver", dtype=int
)

if "grip_condition" in df.columns:
    grip = df["grip_condition"].fillna("UNKNOWN").astype(str)
    grip_dummies = pd.get_dummies(grip, prefix="grip", dtype=int)
else:
    grip_dummies = pd.DataFrame(index=df.index)

X = pd.concat(
    [X_num, compound_dummies, driver_dummies, grip_dummies], axis=1
)
y = df["lap_duration"].reset_index(drop=True)
laps = pd.to_numeric(df["lap_number"], errors="coerce").reset_index(drop=True)

# 5. Temporal split: earlier laps train, later laps test
valid_lap = laps.notna().to_numpy()
X = X.iloc[np.flatnonzero(valid_lap)].reset_index(drop=True)
y = y.iloc[np.flatnonzero(valid_lap)].reset_index(drop=True)
laps = laps.iloc[np.flatnonzero(valid_lap)].reset_index(drop=True)
df = df.iloc[np.flatnonzero(valid_lap)].reset_index(drop=True)

split_lap = int(laps.quantile(0.70))
train_idx = np.flatnonzero((laps <= split_lap).to_numpy())
test_idx = np.flatnonzero((laps > split_lap).to_numpy())

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

print(f"Rows: {len(df)} | Train: {len(train_idx)} | Test: {len(test_idx)}")
print(f"Train laps: {laps.iloc[train_idx].min()}–{laps.iloc[train_idx].max()}")
print(f"Test laps: {laps.iloc[test_idx].min()}–{laps.iloc[test_idx].max()}")
print("Compounds:", sorted(df["compound"].unique()))

# 6. Models
models = {
    "Linear Regression": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        LinearRegression(),
    ),
    "Ridge Regression": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        Ridge(alpha=10.0),
    ),
    "Random Forest": make_pipeline(
        SimpleImputer(strategy="median"),
        RandomForestRegressor(
            n_estimators=400,
            min_samples_leaf=4,
            random_state=42,
            n_jobs=-1,
        ),
    ),
    "XGBoost": make_pipeline(
        SimpleImputer(strategy="median"),
        XGBRegressor(
            n_estimators=300,
            max_depth=4,
            learning_rate=0.04,
            subsample=0.85,
            colsample_bytree=0.85,
            objective="reg:squarederror",
            random_state=42,
            n_jobs=-1,
        ),
    ),
}

# Baseline: median training lap time
baseline = np.full(len(y_test), y_train.median())

results = []
predictions = pd.DataFrame({
    "lap_number": laps.iloc[test_idx].to_numpy(),
    "driver_number": df.iloc[test_idx]["driver_number"].to_numpy(),
    "compound": df.iloc[test_idx]["compound"].to_numpy(),
    "actual": y_test.to_numpy(),
})

for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    predictions[name] = pred
    results.append({
        "model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "Bias": np.mean(pred - y_test.to_numpy()),
        "R2": r2_score(y_test, pred),
    })

results.append({
    "model": "Training median baseline",
    "MAE": mean_absolute_error(y_test, baseline),
    "RMSE": np.sqrt(mean_squared_error(y_test, baseline)),
    "Bias": np.mean(baseline - y_test.to_numpy()),
    "R2": r2_score(y_test, baseline),
})

results_df = pd.DataFrame(results).sort_values("MAE")

print("\n===== FAIR TEMPORAL MODEL COMPARISON =====")
print(results_df.round(4).to_string(index=False))

# Keep predictions and models available for Cells 3 and 4
backtest_predictions = predictions.copy()
backtest_models = models
backtest_split_lap = split_lap
backtest_train_idx = train_idx
backtest_test_idx = test_idx
```

```python

import numpy as np
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/processed/9558")
SESSION = 9558

state = pd.read_parquet(DATA_DIR / "race_state.parquet")
race = state[state["session_key"] == SESSION].copy()

race["compound"] = race["compound"].astype("string").str.upper()
race["lap_number"] = pd.to_numeric(race["lap_number"], errors="coerce")
race["tyre_age"] = pd.to_numeric(race["tyre_age"], errors="coerce")
race["position"] = pd.to_numeric(race["position"], errors="coerce")

race = race.sort_values(
    ["driver_number", "lap_number"]
).reset_index(drop=True)

# Next recorded state for each driver
by_driver = race.groupby("driver_number", sort=False)

race["next_lap"] = by_driver["lap_number"].shift(-1)
race["next_compound"] = by_driver["compound"].shift(-1)
race["next_tyre_age"] = by_driver["tyre_age"].shift(-1)
race["next_position"] = by_driver["position"].shift(-1)
race["next_is_pit_out"] = (
    by_driver["is_pit_out_lap"].shift(-1).fillna(False).astype(bool)
)

# Flag a likely stop between this row and the next recorded row.
# A gap in recorded laps alone is not enough to prove a pit stop.
age_reset = (
    race["next_tyre_age"].notna()
    & race["tyre_age"].notna()
    & (race["next_tyre_age"] < race["tyre_age"])
)

compound_change = (
    race["next_compound"].notna()
    & race["compound"].notna()
    & (race["next_compound"] != race["compound"])
)

pit_event_mask = (
    race["next_is_pit_out"]
    | age_reset
    | compound_change
)

pit_events = race.loc[pit_event_mask, [
    "driver_number",
    "lap_number",
    "position",
    "compound",
    "tyre_age",
    "rainfall",
    "grip_condition",
    "next_lap",
    "next_position",
    "next_compound",
    "next_tyre_age",
    "next_is_pit_out",
]].copy()

pit_events = pit_events.rename(columns={
    "lap_number": "decision_lap",
    "position": "position_before",
    "compound": "compound_before",
    "tyre_age": "tyre_age_before",
    "next_lap": "next_recorded_lap",
    "next_position": "position_next_recorded_lap",
    "next_compound": "compound_after",
    "next_tyre_age": "tyre_age_after",
})

pit_events["position_change_next_recorded_lap"] = (
    pit_events["position_before"]
    - pit_events["position_next_recorded_lap"]
)

print("===== OBSERVED PIT-EVENT CANDIDATES =====")
print("Events found:", len(pit_events))

print(pit_events.head(25).round(2).to_string(index=False))

print("\n===== EVENTS BY COMPOUND TRANSITION =====")
if not pit_events.empty:
    print(
        pit_events.groupby(
            ["compound_before", "compound_after"],
            dropna=False,
        ).size().sort_values(ascending=False).to_string()
    )

# Save for the next strategy-replay stage
output_path = DATA_DIR / "historical_pit_events.parquet"
pit_events.to_parquet(output_path, index=False)

print(f"\nSaved event candidates to: {output_path}")
```

```python

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ============================================================
# CELL 7: ROLLING-ORIGIN TEMPORAL BACKTEST
# ============================================================

# Reuse the feature matrix and target from Cell 2.
# X and y should have the same row order as df.
assert len(X) == len(y) == len(df), (
    f"Length mismatch: X={len(X)}, y={len(y)}, df={len(df)}"
)

backtest_df = df[["lap_number", "compound", "rainfall", "lap_duration"]].copy()
backtest_df = backtest_df.reset_index(drop=True)

X_bt = X.reset_index(drop=True).copy()
y_bt = pd.Series(np.asarray(y), name="lap_duration")

# Keep rows with finite target values.
valid = np.isfinite(pd.to_numeric(y_bt, errors="coerce"))
backtest_df = backtest_df.loc[valid].reset_index(drop=True)
X_bt = X_bt.loc[valid].reset_index(drop=True)
y_bt = y_bt.loc[valid].reset_index(drop=True)

# Chronological validation windows.
# Each fold trains only on earlier laps and predicts a later block.
folds = [
    (15, 16, 24),
    (24, 25, 32),
    (32, 33, 40),
    (40, 41, 46),
    (46, 47, 52),
]

# Reuse the pipelines from Cell 2.
models_bt = {
    name: clone(model)
    for name, model in backtest_models.items()
    if name in ["Linear Regression", "Ridge Regression",
                "Random Forest", "XGBoost"]
}

prediction_rows = []

for train_end, test_start, test_end in folds:
    train_mask = backtest_df["lap_number"] <= train_end
    test_mask = (
        (backtest_df["lap_number"] >= test_start)
        & (backtest_df["lap_number"] <= test_end)
    )

    X_train_fold = X_bt.loc[train_mask]
    y_train_fold = y_bt.loc[train_mask]
    X_test_fold = X_bt.loc[test_mask]
    y_test_fold = y_bt.loc[test_mask]

    if len(X_train_fold) == 0 or len(X_test_fold) == 0:
        print(f"Skipping fold {train_end}->{test_start}-{test_end}: empty split")
        continue

    print(
        f"Fold train laps <= {train_end}, "
        f"test laps {test_start}-{test_end}: "
        f"{len(X_train_fold)} train rows, {len(X_test_fold)} test rows"
    )

    # Baseline: median lap time from the training period.
    baseline_pred = np.full(
        len(y_test_fold),
        float(y_train_fold.median())
    )

    for actual_idx, prediction in zip(y_test_fold.index, baseline_pred):
        prediction_rows.append({
            "fold": f"{test_start}-{test_end}",
            "model": "Training median baseline",
            "row_idx": actual_idx,
            "actual": float(y_bt.loc[actual_idx]),
            "predicted": float(prediction),
        })

    # Fit and evaluate each model.
    for model_name, model in models_bt.items():
        model.fit(X_train_fold, y_train_fold)
        predictions = model.predict(X_test_fold)

        for actual_idx, prediction in zip(y_test_fold.index, predictions):
            prediction_rows.append({
                "fold": f"{test_start}-{test_end}",
                "model": model_name,
                "row_idx": actual_idx,
                "actual": float(y_bt.loc[actual_idx]),
                "predicted": float(prediction),
            })

rolling_predictions = pd.DataFrame(prediction_rows)

rolling_predictions = rolling_predictions.merge(
    backtest_df[["lap_number", "compound", "rainfall"]],
    left_on="row_idx",
    right_index=True,
    how="left",
    validate="many_to_one"
)

rolling_predictions["error"] = (
    rolling_predictions["predicted"] - rolling_predictions["actual"]
)
rolling_predictions["abs_error"] = rolling_predictions["error"].abs()
rolling_predictions["squared_error"] = rolling_predictions["error"] ** 2

# ============================================================
# OVERALL ROLLING BACKTEST RESULTS
# ============================================================

overall_results = (
    rolling_predictions.groupby("model")
    .apply(
        lambda g: pd.Series({
            "n": len(g),
            "MAE": mean_absolute_error(g["actual"], g["predicted"]),
            "RMSE": np.sqrt(mean_squared_error(g["actual"], g["predicted"])),
            "Bias": g["error"].mean(),
        }),
        include_groups=False
    )
    .sort_values("MAE")
)

print("\n===== OVERALL ROLLING TEMPORAL BACKTEST =====")
print(overall_results.round(3).to_string())

# ============================================================
# PERFORMANCE BY TEST WINDOW
# ============================================================

fold_results = (
    rolling_predictions.groupby(["fold", "model"])
    .apply(
        lambda g: pd.Series({
            "n": len(g),
            "MAE": mean_absolute_error(g["actual"], g["predicted"]),
            "RMSE": np.sqrt(mean_squared_error(g["actual"], g["predicted"])),
            "Bias": g["error"].mean(),
        }),
        include_groups=False
    )
    .reset_index()
)

print("\n===== PERFORMANCE BY TEMPORAL FOLD =====")
print(
    fold_results.sort_values(["fold", "MAE"])
    .round(3)
    .to_string(index=False)
)

# ============================================================
# PERFORMANCE BY COMPOUND
# ============================================================

compound_results = (
    rolling_predictions.groupby(["model", "compound"])
    .apply(
        lambda g: pd.Series({
            "n": len(g),
            "MAE": mean_absolute_error(g["actual"], g["predicted"]),
            "RMSE": np.sqrt(mean_squared_error(g["actual"], g["predicted"])),
            "Bias": g["error"].mean(),
        }),
        include_groups=False
    )
    .reset_index()
)

print("\n===== PERFORMANCE BY COMPOUND =====")
print(
    compound_results.sort_values(["compound", "MAE"])
    .round(3)
    .to_string(index=False)
)

# ============================================================
# PERFORMANCE BY WEATHER
# ============================================================

rolling_predictions["weather_regime"] = np.where(
    rolling_predictions["rainfall"].fillna(0).astype(float) > 0,
    "Rain",
    "No rain"
)

weather_results = (
    rolling_predictions.groupby(["model", "weather_regime"])
    .apply(
        lambda g: pd.Series({
            "n": len(g),
            "MAE": mean_absolute_error(g["actual"], g["predicted"]),
            "RMSE": np.sqrt(mean_squared_error(g["actual"], g["predicted"])),
            "Bias": g["error"].mean(),
        }),
        include_groups=False
    )
    .reset_index()
)

print("\n===== PERFORMANCE BY RAINFALL FLAG =====")
print(
    weather_results.sort_values(["weather_regime", "MAE"])
    .round(3)
    .to_string(index=False)
)

# Save predictions and summary tables for the strategy work.
rolling_predictions.to_parquet(
    DATA_DIR / "rolling_temporal_predictions.parquet",
    index=False
)
fold_results.to_csv(DATA_DIR / "rolling_temporal_fold_results.csv", index=False)
compound_results.to_csv(DATA_DIR / "rolling_temporal_compound_results.csv", index=False)
weather_results.to_csv(DATA_DIR / "rolling_temporal_weather_results.csv", index=False)

print("\nSaved rolling temporal backtest outputs to:", DATA_DIR)
```
