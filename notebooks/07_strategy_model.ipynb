```python
fuel_coefficients = (
    fuel_features[fuel_features["lap_number"] > 1]
    .assign(
        fuel_coeff=lambda x:
            x["fuel_effect"] / (x["lap_number"] - 1)
    )
    .groupby("driver_number")["fuel_coeff"]
    .median()
)

print(fuel_coefficients)
```

```python
def estimate_fuel_effect(driver_number, lap_number):
    """
    Estimate cumulative fuel effect for a driver at a given lap.
    Used only by the strategy simulator.
    """

    if driver_number not in fuel_coefficients.index:
        return np.nan

    coefficient = fuel_coefficients.loc[driver_number]

    return coefficient * (lap_number - 1)

print("Driver 44:")
for lap in [1, 10, 27, 30, 40, 52]:
    print(lap, estimate_fuel_effect(44, lap))
```

```python
def build_model_input(state):
    row = pd.DataFrame([{
        "driver_number": state["driver_number"],
        "compound": state["compound"],
        "tyre_age": state["tyre_age"],
        "fuel_effect": state["fuel_effect_sim"],
        "traffic_ahead": state["traffic_ahead"],
        "traffic_behind": state["traffic_behind"],
        "position_change": state["position_change"],
        "track_evolution": state["track_evolution"],
        "rainfall": state["rainfall"],
        "humidity": state["humidity"],
        "air_temperature": state["air_temperature"],
        "wind_speed": state["wind_speed"],
    }])

    row = pd.get_dummies(
        row,
        columns=["driver_number", "compound"],
        drop_first=True
    )

    row = row.reindex(
        columns=pace_features,
        fill_value=0
    )

    return row
```

```python
def predict_lap_time(lap_number, driver_number):

    # Get the strategy state directly
    rows = strategy_state[
        (strategy_state["driver_number"] == driver_number) &
        (strategy_state["lap_number"] == lap_number)
    ]

    if rows.empty:
        raise ValueError(
            f"No strategy state found for driver {driver_number}, "
            f"lap {lap_number}"
        )

    state = rows.iloc[0]

    # Check required inputs
    required = [
        "tyre_age",
        "compound",
        "traffic_ahead",
        "traffic_behind",
        "position_change",
        "track_evolution",
        "rainfall",
        "humidity",
        "air_temperature",
        "wind_speed",
        "fuel_effect_sim"
    ]

    missing = [
        col for col in required
        if col not in state.index or pd.isna(state[col])
    ]

    if missing:
        raise ValueError(
            f"Missing model inputs: {missing}"
        )

    # Build model input
    X = pd.DataFrame([{
        "driver_number": state["driver_number"],
        "compound": state["compound"],
        "tyre_age": state["tyre_age"],
        "fuel_effect": state["fuel_effect_sim"],
        "traffic_ahead": state["traffic_ahead"],
        "traffic_behind": state["traffic_behind"],
        "position_change": state["position_change"],
        "track_evolution": state["track_evolution"],
        "rainfall": state["rainfall"],
        "humidity": state["humidity"],
        "air_temperature": state["air_temperature"],
        "wind_speed": state["wind_speed"],
    }])

    # Same categorical encoding used during model training
    X = pd.get_dummies(
        X,
        columns=["driver_number", "compound"],
        drop_first=True
    )

    # EXACT feature order expected by XGBoost
    X = X.reindex(
        columns=pace_features,
        fill_value=0
    )

    # Predict
    prediction = pace_model.predict(X)[0]

    return prediction
```

```python
# ============================================
# REBUILD TRAINING-CONSISTENT TRACK EVOLUTION
# ============================================

# Same filtering logic used in the pace notebook
race_pace_data = lap_performance[
    lap_performance["lap_duration"].notna()
    & (lap_performance["is_pit_out_lap"] == False)
    & lap_performance["compound"].isin(
        ["SOFT", "MEDIUM", "HARD"]
    )
].copy()

# Median pace and driver count for each lap
race_pace_reliability = (
    race_pace_data
    .groupby("lap_number")["lap_duration"]
    .agg(
        median_lap_time="median",
        driver_count="count"
    )
    .reset_index()
)

# Ensure every lap exists in the race timeline
full_laps = pd.DataFrame({
    "lap_number": range(
        int(race_pace_reliability["lap_number"].min()),
        int(race_pace_reliability["lap_number"].max()) + 1
    )
})

race_pace_reliability = (
    full_laps
    .merge(
        race_pace_reliability,
        on="lap_number",
        how="left",
        validate="one_to_one"
    )
    .sort_values("lap_number")
    .reset_index(drop=True)
)

# Exclude unreliable lap medians
MIN_DRIVERS = 5

race_pace_reliability["reliable_median_lap_time"] = (
    race_pace_reliability["median_lap_time"]
    .where(
        race_pace_reliability["driver_count"] >= MIN_DRIVERS
    )
)

# Previous lap only
track_evolution_data = race_pace_reliability[
    ["lap_number", "reliable_median_lap_time"]
].copy()

track_evolution_data["previous_pace"] = (
    track_evolution_data["reliable_median_lap_time"].shift(1)
)

# Require 3 consecutive previous race laps
track_evolution_data["recent_field_pace"] = (
    track_evolution_data["previous_pace"]
    .rolling(window=3, min_periods=3)
    .median()
)

# Reference pace from laps 5 to 15
reference_pace = track_evolution_data.loc[
    track_evolution_data["lap_number"].between(5, 15),
    "reliable_median_lap_time"
].median()

# Track evolution
track_evolution_data["track_evolution"] = (
    track_evolution_data["recent_field_pace"] - reference_pace
)

# Lookup for strategy state
track_evolution_lookup = (
    track_evolution_data
    .set_index("lap_number")["track_evolution"]
    .to_dict()
)

print("Reference pace:", round(reference_pace, 3))
print("\nTrack evolution around lap 30:")
print(
    track_evolution_data[
        track_evolution_data["lap_number"].between(25, 35)
    ][
        [
            "lap_number",
            "reliable_median_lap_time",
            "recent_field_pace",
            "track_evolution"
        ]
    ].round(3).to_string(index=False)
)
```

```python
# Start from the original race state
strategy_state = race_state.copy()

# Rebuild track evolution using the training-time logic
race_pace_data = lap_performance[
    lap_performance["lap_duration"].notna()
    & (lap_performance["is_pit_out_lap"] == False)
    & lap_performance["compound"].isin(["SOFT", "MEDIUM", "HARD"])
].copy()

race_pace_reliability = (
    race_pace_data.groupby("lap_number")["lap_duration"]
    .agg(median_lap_time="median", driver_count="count")
    .reset_index()
)

full_laps = pd.DataFrame({
    "lap_number": range(
        int(race_pace_reliability["lap_number"].min()),
        int(race_pace_reliability["lap_number"].max()) + 1
    )
})

race_pace_reliability = (
    full_laps
    .merge(
        race_pace_reliability,
        on="lap_number",
        how="left",
        validate="one_to_one"
    )
    .sort_values("lap_number")
    .reset_index(drop=True)
)

MIN_DRIVERS = 5

race_pace_reliability["reliable_median_lap_time"] = (
    race_pace_reliability["median_lap_time"]
    .where(race_pace_reliability["driver_count"] >= MIN_DRIVERS)
)

track_evolution_data = race_pace_reliability[
    ["lap_number", "reliable_median_lap_time"]
].copy()

track_evolution_data["previous_pace"] = (
    track_evolution_data["reliable_median_lap_time"].shift(1)
)

track_evolution_data["recent_field_pace"] = (
    track_evolution_data["previous_pace"]
    .rolling(window=3, min_periods=3)
    .median()
)

reference_pace = track_evolution_data.loc[
    track_evolution_data["lap_number"].between(5, 15),
    "reliable_median_lap_time"
].median()

track_evolution_data["track_evolution"] = (
    track_evolution_data["recent_field_pace"] - reference_pace
)

track_evolution_lookup = track_evolution_data.set_index(
    "lap_number"
)["track_evolution"].to_dict()

# Merge fuel and traffic features
strategy_state = strategy_state.merge(
    fuel_features,
    on=["driver_number", "lap_number"],
    how="left",
    validate="one_to_one"
)

strategy_state = strategy_state.merge(
    traffic_features.drop_duplicates(
        subset=["driver_number", "lap_number"]
    ),
    on=["driver_number", "lap_number"],
    how="left",
    validate="one_to_one",
    suffixes=("", "_traffic")
)

# Apply training-consistent track evolution
strategy_state["track_evolution"] = (
    strategy_state["lap_number"].map(track_evolution_lookup)
)

print("Strategy state shape:", strategy_state.shape)
print("Duplicate driver/lap keys:", strategy_state.duplicated(
    ["driver_number", "lap_number"]
).sum())
print("Reference pace:", reference_pace)
print(
    "Driver 44, Lap 30 track evolution:",
    strategy_state.loc[
        (strategy_state["driver_number"] == 44)
        & (strategy_state["lap_number"] == 30),
        "track_evolution"
    ].to_list()
)
```

```python
fuel_coefficients = (
    fuel_features.loc[fuel_features["lap_number"] > 1]
    .assign(
        fuel_coeff=lambda x:
        x["fuel_effect"] / (x["lap_number"] - 1)
    )
    .groupby("driver_number")["fuel_coeff"]
    .median()
)

def estimate_fuel_effect(driver_number, lap_number):
    if driver_number not in fuel_coefficients.index:
        return np.nan

    return fuel_coefficients.loc[driver_number] * (lap_number - 1)

strategy_state["fuel_effect_sim"] = strategy_state.apply(
    lambda row: estimate_fuel_effect(
        row["driver_number"],
        row["lap_number"]
    ),
    axis=1
)

print(
    strategy_state[
        ["driver_number", "lap_number", "fuel_effect", "fuel_effect_sim"]
    ].head(10)
)

print("Missing simulated fuel effects:",
      strategy_state["fuel_effect_sim"].isna().sum())
```

```python
# Select rows with complete required inference features
validation_data = strategy_state.loc[
    strategy_state["compound"].isin(["SOFT", "MEDIUM", "HARD"])
    & (~strategy_state["is_pit_out_lap"].fillna(True))
    & strategy_state["lap_duration"].notna()
    & strategy_state["fuel_effect"].notna()
    & strategy_state["traffic_ahead"].notna()
    & strategy_state["traffic_behind"].notna()
    & strategy_state["position_change"].notna()
].copy()

# Match the training-time handling of missing track evolution
validation_data["track_evolution"] = (
    validation_data["track_evolution"].fillna(0.0)
)

# Build model inputs
X_val = validation_data[
    [
        "driver_number",
        "compound",
        "tyre_age",
        "fuel_effect",
        "traffic_ahead",
        "traffic_behind",
        "position_change",
        "track_evolution",
        "rainfall",
        "humidity",
        "air_temperature",
        "wind_speed",
    ]
].copy()

X_val = pd.get_dummies(
    X_val,
    columns=["driver_number", "compound"],
    drop_first=True
)

X_val = X_val.reindex(columns=pace_features, fill_value=0)

# Predict and evaluate
y_true = validation_data["lap_duration"].to_numpy()
y_pred = pace_model.predict(X_val)

rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
mae = np.mean(np.abs(y_true - y_pred))

print("Validation rows:", len(validation_data))
print(f"RMSE: {rmse:.4f} seconds")
print(f"MAE:  {mae:.4f} seconds")

validation_data["predicted_lap_duration"] = y_pred
validation_data["prediction_error"] = y_pred - y_true

print(
    validation_data[
        [
            "driver_number",
            "lap_number",
            "compound",
            "lap_duration",
            "predicted_lap_duration",
            "prediction_error",
        ]
    ].head(10).round(3).to_string(index=False)
)
```

```python
# Absolute prediction error
validation_data["abs_error"] = (
    validation_data["prediction_error"].abs()
)

# 1. Inspect the 15 worst predictions
worst_predictions = validation_data.nlargest(15, "abs_error")

print("15 worst predictions:")
print(
    worst_predictions[
        [
            "driver_number",
            "lap_number",
            "compound",
            "lap_duration",
            "predicted_lap_duration",
            "prediction_error",
            "rainfall",
            "tyre_age",
            "track_evolution",
        ]
    ].round(3).to_string(index=False)
)

# 2. Error distribution
print("\nError distribution:")
print(
    validation_data["prediction_error"]
    .describe(percentiles=[0.50, 0.90, 0.95, 0.99])
    .round(3)
)

# 3. Check error by rainfall condition
print("\nError by rainfall:")
print(
    validation_data.groupby("rainfall", dropna=False)
    .agg(
        laps=("prediction_error", "size"),
        MAE=("abs_error", "mean"),
        mean_error=("prediction_error", "mean"),
    )
    .round(3)
)

# 4. Check error by compound
print("\nError by compound:")
print(
    validation_data.groupby("compound")
    .agg(
        laps=("prediction_error", "size"),
        MAE=("abs_error", "mean"),
        mean_error=("prediction_error", "mean"),
    )
    .round(3)
)
```

```python
# Flag start laps
validation_data["is_lap_1"] = validation_data["lap_number"].eq(1)

def summarize_errors(df, label):
    if df.empty:
        print(f"{label}: no rows")
        return

    errors = df["prediction_error"]
    print(
        f"{label}: n={len(df)}, "
        f"MAE={errors.abs().mean():.3f}s, "
        f"RMSE={np.sqrt((errors ** 2).mean()):.3f}s, "
        f"Bias={errors.mean():.3f}s"
    )

summarize_errors(validation_data, "All laps")
summarize_errors(
    validation_data[validation_data["lap_number"] > 1],
    "After Lap 1"
)
summarize_errors(
    validation_data[
        (validation_data["lap_number"] > 1)
        & (validation_data["rainfall"] == 0)
    ],
    "After Lap 1, no rainfall"
)
summarize_errors(
    validation_data[
        (validation_data["lap_number"] > 1)
        & (validation_data["rainfall"] > 0)
    ],
    "After Lap 1, rainfall"
)

print("\nMean absolute error by lap number:")
print(
    validation_data.groupby("is_lap_1")["prediction_error"]
    .agg(
        laps="count",
        mean_error="mean",
        MAE=lambda x: x.abs().mean(),
    )
    .round(3)
)  
```
