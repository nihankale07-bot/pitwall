```python
import pandas as pd
from pathlib import Path

POSITION_PATH = Path("../data/processed/9558/position.parquet")

position = pd.read_parquet(POSITION_PATH)

print("Shape:", position.shape)
print("\nColumns:")
print(position.columns.tolist())

print("\nFirst 10 rows:")
display(position.head(10))

print("\nUnique drivers:", position["driver_number"].nunique())

print("\nRows per driver:")
display(
    position.groupby("driver_number")
    .size()
    .sort_values()
)

print("\nTimestamp range:")
print("Start:", position["date"].min())
print("End:  ", position["date"].max())
```

```python
print("Unique drivers:", position["driver_number"].nunique())

print("Rows per driver:")
display(
    position.groupby("driver_number")
    .size()
    .sort_values()
)
print("Timestamp range:")
print("Start:", position["date"].min())
print("End:  ", position["date"].max())
```

```python
display(
    position[
        ["date", "driver_number", "position"]
    ].sort_values("date").head(30)
)
print(position.dtypes)

print("\nPosition values:")
print(sorted(position["position"].dropna().unique()))

print("\nMissing values:")
print(position[["date", "driver_number", "position"]].isna().sum())
```

```python
position_events = (
    position.sort_values("date").groupby("date").agg(
        cars_reporting=("driver_number", "nunique"),
        min_position=("position", "min"),
        max_position=("position", "max")
    ).reset_index()
)

display(position_events.head(20))
print("Total position events:", len(position_events))
print("Cars report per event:")

display(
    position_events["cars_reporting"].value_counts().sort_index()
)

display(
    position_events.sort_values("cars_reporting").head(20)
)
```

```python
position_times = (
    position[["date"]].drop_duplicates().sort_values("date").reset_index(drop=True)
)

print("Unique position timestamps:", len(position_times))
display(position_times.head(10))
```

```python
drivers = (
    position["driver_number"].drop_duplicates().sort_values().tolist()
)

position_timeline = pd.MultiIndex.from_product(
    [position_times["date"], drivers],
    names=["date", "driver_number"]
).to_frame(index=False)

print("Timeline shape:", position_timeline.shape)
display(position_timeline.head(20))
```

```python
position_timeline = position_timeline.merge(
    position[["date", "driver_number", "position"]],
    on=["date","driver_number"],
    how="left"
)

print("Shape:", position_timeline.shape)

display(
    position_timeline.head(25)
)
```

```python
position_timeline = (
    position_timeline.sort_values(["driver_number", "date"]).copy()
)

position_timeline["position"] = (
    position_timeline.groupby("driver_number")["position"].ffill()
)

display(
    position_timeline.head(25)
)
```

```python
print("Missing positions:", position_timeline["position"].isna().sum())

print("\nRows per timestamp:")
display(
    position_timeline
    .groupby("date")
    .size()
    .value_counts()
    .sort_index()
)
```

```python
position_timeline["position_update"] = (
    position_timeline["position"].ne(
        position_timeline.groupby("driver_number")["position"].shift()
    )
)
position_timeline["last_position_update"] = (
    position_timeline["date"].where(position_timeline["position_update"])
    .groupby(position_timeline["driver_number"]).ffill()

)

position_timeline["seconds_since_position_update"] = (
    position_timeline["date"] - position_timeline["last_position_update"]
).dt.total_seconds()

display(
    position_timeline[
        [
            "date",
            "driver_number",
            "position",
            "last_position_update",
            "seconds_since_position_update"
        ]
    ].head(30)
)
```

```python
stale_position = (
    position_timeline[
        [
            "driver_number",
            "date",
            "position",
            "seconds_since_position_update"
        ]
    ].sort_values("seconds_since_position_update", ascending=False)
)
display(stale_position.head(30))
```

```python
position_events= position[
    ["date", "driver_number", "position"]
].copy()

position_events["position_event"] = True

display(position_events.head(20))
```

```python
position_timeline = position_timeline.drop(
    columns=["position_event", "last_position_update", "seconds_since_position_update"],
    errors="ignore"
)

position_timeline = position_timeline.merge(
    position_events,
    on=["date", "driver_number", "position"],
    how="left"
)

position_timeline["position_event"] = (
    position_timeline["position_event"].fillna(False)
)

print("Actual position events:", position_timeline["position_event"].sum())

display(
    position_timeline[
        ["date", "driver_number", "position", "position_event"]
    ].head(30)
)
```

```python
position_timeline["last_position_update"] = (
    position_timeline["date"]
    .where(position_timeline["position_event"])
    .groupby(position_timeline["driver_number"])
    .ffill()
)

position_timeline["seconds_since_position_update"] = (
    position_timeline["date"]
    - position_timeline["last_position_update"]
).dt.total_seconds()

display(
    position_timeline[
        [
            "driver_number",
            "date",
            "position",
            "position_event",
            "seconds_since_position_update"
        ]
    ]
    .sort_values("seconds_since_position_update", ascending=False)
    .head(30)
)
```

```python
from pathlib import Path
import pandas as pd

LAP_PATH = Path("../data/processed/9558/lap_performance.parquet")

lap_performance = pd.read_parquet(LAP_PATH)

print("Shape:", lap_performance.shape)
print("Columns:")
print(lap_performance.columns.tolist())
```

```python
lap_timeline = (
    lap_performance[
        [
            "date_start",
            "driver_number",
            "lap_number",
        ]
    ]
    .dropna(subset=["date_start"])
    .sort_values(["driver_number","date_start"])
    .reset_index(drop=True)
)
print("Shape:", lap_timeline.shape)
print(lap_timeline.head(10))
```

```python
position_events = (
    position[
        [
            "date",
            "driver_number",
            "position",
        ]
    ]
    .dropna(subset=["date", "position"])
    .sort_values(["driver_number", "date"])
    .reset_index(drop=True)
)

print("Shape:", position_events.shape)
print(position_events.head(20))
```

```python
lap_timeline_asof = (
    lap_timeline.sort_values(["date_start", "driver_number"]).reset_index(drop=True)
)

position_events_asof = (
    position_events
    .rename(columns={"date":"position_event_time"})
    .sort_values(["position_event_time","driver_number"])
    .reset_index(drop=True)
)

lap_with_position = pd.merge_asof(
    lap_timeline_asof,
    position_events_asof,
    left_on="date_start",
    right_on="position_event_time",
    by="driver_number",
    direction="backward"
)

print("Shape:", lap_with_position.shape)
print(lap_with_position.head(20))
```

```python
lap_with_position["position_age_seconds"] = (
    lap_with_position["date_start"]
    - lap_with_position["position_event_time"]
).dt.total_seconds()

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "date_start",
            "position_event_time",
            "position",
            "position_age_seconds",
        ]
    ].head(30)
)

print(lap_with_position["position_age_seconds"].describe())
```

```python
race_start = lap_performance["date_start"].min()

print("Race start:", race_start)

lap_with_position["is_race_started"] = (
    lap_with_position["date_start"] >= race_start
)

lap_with_position["position_age_seconds"] = (
    lap_with_position["date_start"]
    - lap_with_position["position_event_time"]
).dt.total_seconds()

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "position",
            "position_event_time",
            "position_age_seconds",
            "is_race_started",
        ]
    ].head(20)
)
```

```python
driver_activity = (
    lap_performance
    .groupby("driver_number")
    .agg(
        first_lap=("lap_number", "min"),
        last_lap=("lap_number", "max"),
        last_lap_time=("date_start", "max"),
        laps_recorded=("lap_number", "count"),
    )
    .reset_index()
)

print(driver_activity)
```

```python
lap_with_position = lap_with_position.merge(
    driver_activity[
        [
            "driver_number",
            "last_lap",
            "last_lap_time",
        ]
    ],
    on="driver_number",
    how="left"
)

lap_with_position["seconds_from_last_lap"] = (
    lap_with_position["last_lap_time"]
    - lap_with_position["date_start"]
).dt.total_seconds()

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "date_start",
            "last_lap",
            "last_lap_time",
            "seconds_from_last_lap",
        ]
    ].tail(20)
)
```

```python
lap_with_position["position_status"] = "ACTIVE"

lap_with_position.loc[
    lap_with_position["position_event_time"] < race_start,
    "position_status"
] = "PRE_RACE"

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "date_start",
            "position",
            "position_event_time",
            "position_age_seconds",
            "position_status",
        ]
    ].head(30)
)
```

```python
race_position_events = position_events[
    position_events["date"] >= race_start
].copy()

race_position_events["seconds_since_position_update"] = (
    race_position_events
    .groupby("driver_number")["date"]
    .diff()
    .dt.total_seconds()
)

print(
    race_position_events["seconds_since_position_update"].describe()
)
```

```python
largest_gaps = (
    race_position_events
    .dropna(subset=["seconds_since_position_update"])
    .sort_values("seconds_since_position_update", ascending=False)
    [
        [
            "driver_number",
            "date",
            "position",
            "seconds_since_position_update",
        ]
    ]
    .head(20)
)

print(largest_gaps)
```

```python
race_position_events = position_events[
    position_events["date"] >= race_start
].copy()

race_position_events["seconds_since_position_update"] = (
    race_position_events
    .groupby("driver_number")["date"]
    .diff()
    .dt.total_seconds()
)

print(
    race_position_events[
        race_position_events["seconds_since_position_update"] > 600
    ][
        [
            "driver_number",
            "date",
            "position",
            "seconds_since_position_update",
        ]
    ]
    .sort_values("seconds_since_position_update", ascending=False)
    .to_string(index=False)
)
```

```python
lap_with_position["driver_active"] = (
    lap_with_position["lap_number"]
    <= lap_with_position["last_lap"]
)

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "last_lap",
            "driver_active",
        ]
    ].tail(20)
)
```

```python
lap_with_position = (
    lap_with_position
    .sort_values(["driver_number", "lap_number"])
    .reset_index(drop=True)
)

lap_with_position["position_change"] = (
    lap_with_position
    .groupby("driver_number")["position"]
    .diff()
)

lap_with_position["positions_gained"] = (
    -lap_with_position["position_change"]
).clip(lower=0)

lap_with_position["positions_lost"] = (
    lap_with_position["position_change"]
).clip(lower=0)

max_position = lap_with_position[
    lap_with_position["driver_number"] == 1
][
    [
        "lap_number",
        "date_start",
        "position",
        "position_change",
        "positions_gained",
        "positions_lost",
        "position_event_time",
        "position_age_seconds"
    ]
]

print(max_position.to_string(index=False))
```

```python
lap_with_position["position_change"] = (
    lap_with_position["position_change"]
    .fillna(0)
)

lap_with_position["positions_gained"] = (
    lap_with_position["positions_gained"]
    .fillna(0)
)

lap_with_position["positions_lost"] = (
    lap_with_position["positions_lost"]
    .fillna(0)
)

lap_with_position["position_change_last_lap"] = (
    lap_with_position
    .groupby("driver_number")["position_change"]
    .shift(1)
)

lap_with_position["positions_gained_last_lap"] = (
    lap_with_position
    .groupby("driver_number")["positions_gained"]
    .shift(1)
)

lap_with_position["positions_lost_last_lap"] = (
    lap_with_position
    .groupby("driver_number")["positions_lost"]
    .shift(1)
)

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "position",
            "position_change",
            "positions_gained",
            "positions_lost",
            "position_change_last_lap",
            "positions_gained_last_lap",
            "positions_lost_last_lap",
        ]
    ].head(25)
)
```

```python
lap_with_position["driver_ahead"] = (
    lap_with_position["position"] - 1
)

lap_with_position["driver_behind"] = (
    lap_with_position["position"] + 1
)

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "position",
            "driver_ahead",
            "driver_behind"
        ]
    ].head(10)
)
```

```python
ahead_lookup = (
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "position"
        ]
    ]
    .rename(
        columns={
            "driver_number": "driver_ahead_number",
            "position": "driver_ahead_position"
        }
    )
)
lap_with_position = lap_with_position.merge(
    ahead_lookup,
    left_on=["lap_number", "driver_ahead"],
    right_on=["lap_number", "driver_ahead_position"],
    how="left"
)

lap_with_position = lap_with_position.drop(
    columns=["driver_ahead_position"]
)

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "position",
            "driver_ahead",
            "driver_ahead_number",
            "driver_behind"
        ]
    ].head(10)
)
```

```python
# Remove previously added traffic-performance columns
traffic_columns = [
    "ahead_lap_duration",
    "ahead_compound",
    "ahead_tyre_age",
    "ahead_stint_number",
    "behind_lap_duration",
    "behind_compound",
    "behind_tyre_age",
    "behind_stint_number",
    "driver_behind_number",
]

lap_with_position = lap_with_position.drop(
    columns=traffic_columns,
    errors="ignore"
)

ahead_performance = (
    lap_performance[
        [
            "driver_number",
            "lap_number",
            "lap_duration",
            "compound",
            "tyre_age",
            "stint_number"
        ]
    ]
    .rename(
        columns={
            "driver_number": "driver_ahead_number",
            "lap_duration": "ahead_lap_duration",
            "compound": "ahead_compound",
            "tyre_age": "ahead_tyre_age",
            "stint_number": "ahead_stint_number"
        }
    )
)

behind_performance = (
    lap_performance[
        [
            "driver_number",
            "lap_number",
            "lap_duration",
            "compound",
            "tyre_age",
            "stint_number"
        ]
    ]
    .rename(
        columns={
            "driver_number": "driver_behind_number",
            "lap_duration": "behind_lap_duration",
            "compound": "behind_compound",
            "tyre_age": "behind_tyre_age",
            "stint_number": "behind_stint_number"
        }
    )
)

lap_with_position = lap_with_position.merge(
    ahead_performance,
    on=["driver_ahead_number", "lap_number"],
    how="left"
)

lap_with_position = lap_with_position.merge(
    behind_performance,
    left_on=["lap_number", "driver_behind"],
    right_on=["lap_number", "driver_behind_number"],
    how="left"
)

print(
    lap_with_position[
        [
            "driver_number",
            "lap_number",
            "position",
            "driver_ahead_number",
            "ahead_lap_duration",
            "ahead_compound",
            "ahead_tyre_age",
            "driver_behind_number",
            "behind_lap_duration",
            "behind_compound",
            "behind_tyre_age"
        ]
    ].head(25)
)
```

```python
max_traffic = lap_with_position[
    lap_with_position["driver_number"] == 1
][
    [
        "lap_number",
        "position",
        "driver_ahead",
        "driver_ahead_number",
        "driver_behind",
        "driver_behind_number"
    ]
]

print(max_traffic.to_string(index=False))
```

```python
traffic_features = lap_with_position[
    [
        "driver_number",
        "lap_number",
        "position",
        "driver_ahead_number",
        "driver_behind_number",
        "position_change"
    ]
].copy()

traffic_features["traffic_ahead"] = (
    traffic_features["driver_ahead_number"].notna()
).astype(int)

traffic_features["traffic_behind"] = (
    traffic_features["driver_behind_number"].notna()
).astype(int)

traffic_features = traffic_features[
    [
        "driver_number",
        "lap_number",
        "position",
        "traffic_ahead",
        "traffic_behind",
        "position_change"
    ]
]

print("Traffic features shape:", traffic_features.shape)

print(
    traffic_features.head(20).to_string(index=False)
)
 
```

```python
print("\nMissing values:")
print(traffic_features.isna().sum())

print("\nTraffic ahead distribution:")
print(traffic_features["traffic_ahead"].value_counts())

print("\nTraffic behind distribution:")
print(traffic_features["traffic_behind"].value_counts())

print("\nPosition change distribution:")
print(traffic_features["position_change"].value_counts().sort_index())
```
