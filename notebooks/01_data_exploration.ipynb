```python
import requests
```

```python
url = "https://api.openf1.org/v1/sessions"

response = requests.get(url)

print(response.status_code)
```

```python
data = response.json()

print(type(data))
print(len(data))
```

```python
british_2024 = [
    session for session in data
    if session["year"] == 2024
    and session["country_name"] == "United Kingdom"
]

british_2024
```

```python
for session in british_2024:
    print(
        session["session_key"],
        "|",
        session["session_name"],
        "|",
        session["session_type"]
    )
```

```python
race_session = [
    session for session in british_2024
    if session["session_type"] == "Race"
][0]

race_session
```

```python
laps_url = "https://api.openf1.org/v1/laps"

params = {
    "session_key": race_session["session_key"],
    
}

response = requests.get(laps_url, params=params)

print(response.status_code)
```

```python
laps = response.json()

print(type(laps))
print(len(laps))

laps[:5]
```

```python
print("Total lap records:", len(laps))

drivers = sorted(set(lap["driver_number"] for lap in laps))
print("Drivers:", drivers)

lap_numbers = [lap["lap_number"] for lap in laps]
print("Minimum lap:", min(lap_numbers))
print("Maximum lap:", max(lap_numbers))
```

```python
from collections import Counter

lap_counts = Counter(lap["driver_number"] for lap in laps)

for driver, count in sorted(lap_counts.items()):
    print(f"Driver {driver}: {count} records")
```

```python
positions_url = "https://api.openf1.org/v1/position"
params = {"session_key": race_session["session_key"]}
response = requests.get(positions_url, params=params)
print(f"Status code: {response.status_code}")
if response.status_code == 200:
    positions = response.json()
    print("Data fetched")
else:
    positions = None
    print(f"Failed to retrieve data. Error: {response.text}")
```

```python
print(type(positions))
print(len(positions))
print(positions[:3])
```

```python
for position in positions[:20]:
    print(
        position["date"],
        "| Driver:", position["driver_number"],
        "| Position:", position["position"],
    )
```

```python
print("Total drivers: ", len(drivers))
print("Available fields:")
print("Content of drivers: ",drivers)
print(drivers[0])
print("First driver:", drivers[0])
```

```python
driver_lookup = {
    driver["driver_number"]: driver["full_name"]
    for driver in drivers
}

print(driver_lookup)
```

```python
stints_url = "https://api.openf1.org/v1/stints"

params = {
    "session_key": race_session["session_key"]
}

response = requests.get(stints_url, params=params)

print("Status code:", response.status_code)

stints = response.json()

print("Type:", type(stints))
print("Total stints:", len(stints))
print(stints[:5])
```

```python
ver_stints = [
    stint for stint in stints
    if stint["driver_number"] == 1
]

for stint in ver_stints:
    print(
         "Stint:", stint["stint_number"],
        "| Compound:", stint["compound"],
        "| Start:", stint["lap_start"],
        "| End:", stint["lap_end"],
        "| Tyre age:", stint["tyre_age_at_start"]
    )
```

```python
for stint in stints:
    if stint["lap_start"] is None or stint["lap_end"] is None:
        print(stint)
```

```python
ver_laps = [
    lap for lap in laps
    if lap["driver_number"] == 1
]

print("Max Verstappen's lap records:", len(ver_laps))

for lap in ver_laps[:10]:
    print(
        "Lap:", lap["lap_number"],
        "| Time:", lap["lap_duration"],
        "| Sector 1:", lap["duration_sector_1"],
        "| Sector 2:", lap["duration_sector_2"],
        "| Sector 3:", lap["duration_sector_3"]
    )
```

```python
ver_positions = [
    position for position in positions
    if position["driver_number"] == 1
]

print("Max Verstappen's position records:", len(ver_positions))

for position in ver_positions[:20]:
    print(
        position["date"],
        "| Position:", position["position"]
    )
```
